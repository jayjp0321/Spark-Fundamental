<h2 style="color:#1d4ed8;font-size:21px;border-left:5px solid #3b82f6;padding:10px 12px;background:#eff6ff;border-radius:5px;">1.1. What Is Apache Spark?</h2>
<p style="font-size:14px;line-height:1.85;"><strong style="color:#1d4ed8;">Answer:</strong> Apache Spark is an open-source, distributed data-processing engine designed to process large-scale datasets across multiple CPU cores and, when deployed on a cluster, across multiple machines.</p>
<p style="font-size:14px;line-height:1.85;">It provides programming APIs for Python (PySpark), Scala, Java and R. Developers express data-processing operations through these APIs, while Spark plans and coordinates their execution.</p>
<p style="font-size:14px;line-height:1.85;">Spark supports execution planning, job and task scheduling, partition processing, data shuffling and fault recovery. Its ecosystem also supports SQL analytics, structured streaming and machine learning workloads.</p>
<div style="box-sizing:border-box;background:#ecfdf5;border-left:4px solid #10b981;padding:14px 16px;border-radius:7px;margin:16px 0;">
<h3 style="color:#047857;margin:0 0 7px;font-size:16px;">💡 In Other Words</h3>
<p style="margin:0;line-height:1.85;font-size:14px;">Think of Spark as the engine that executes data-processing work and the platform that provides APIs through which we describe that work. For example, we write PySpark code, and Spark plans and executes the corresponding operations.</p>
</div>
<h2 style="color:#6d28d9;font-size:21px;border-left:5px solid #8b5cf6;padding:10px 12px;background:#f5f3ff;border-radius:5px;margin-top:26px;">1.2. Why Is Spark Used?</h2>
<h3 style="color:#7c3aed;font-size:16px;line-height:1.6;">What problem is Spark designed to solve that Pandas isn't designed to solve?</h3>
<p style="font-size:14px;line-height:1.85;"><strong style="color:#6d28d9;">Answer:</strong> As datasets grow, processing them can become computationally expensive and memory-intensive. Pandas is highly effective for single-machine data analysis, but its practical limits depend on that machine's resources and the operations being performed.</p>
<p style="font-size:14px;line-height:1.85;">Spark addresses large-scale workloads by dividing data into partitions and executing tasks in parallel across available CPU cores and, in cluster deployments, across multiple machines. This allows a workload to use aggregate cluster resources instead of relying entirely on one machine.</p>
<p style="font-size:14px;line-height:1.85;">Spark also supports fault recovery and distributed execution planning, making it useful for large-scale data engineering, analytics and machine learning pipelines.</p>

<h2 style="color:#c2410c;font-size:21px;border-left:5px solid #f59e0b;padding:10px 12px;background:#fff7ed;border-radius:5px;margin-top:26px;">1.3. Important Distinction: Lazy Evaluation</h2>
<p style="font-size:14px;line-height:1.85;">Spark supports <strong>lazy evaluation</strong>, but this does not mean Spark is called lazy because it processes data in chunks.</p>
<div style="border:1px solid #fed7aa;border-radius:8px;overflow:hidden;margin:14px 0;">
<div style="box-sizing:border-box;padding:12px 14px;background:#fff7ed;border-bottom:1px solid #fed7aa;"><strong style="color:#9a3412;">1. Partitioning</strong><p style="margin:4px 0 0;font-size:14px;line-height:1.7;">Divides a dataset into logical units of work.</p></div>
<div style="box-sizing:border-box;padding:12px 14px;background:#ffffff;border-bottom:1px solid #fed7aa;"><strong style="color:#9a3412;">2. Parallelism</strong><p style="margin:4px 0 0;font-size:14px;line-height:1.7;">Allows multiple tasks to execute concurrently when resources are available.</p></div>
<div style="box-sizing:border-box;padding:12px 14px;background:#fff7ed;"><strong style="color:#9a3412;">3. Lazy Evaluation</strong><p style="margin:4px 0 0;font-size:14px;line-height:1.7;">Generally delays execution of transformations until an action requires a result.</p></div>
</div>
<p style="font-size:14px;line-height:1.85;">For example, Spark can record a <code style="background:#f1f5f9;padding:2px 4px;border-radius:3px;">filter()</code> transformation in its execution plan and perform the required computation when an action such as <code style="background:#f1f5f9;padding:2px 4px;border-radius:3px;">count()</code> is invoked.</p>
<div style="box-sizing:border-box;background:linear-gradient(100deg,#eff6ff,#f5f3ff);border:1px solid #c4b5fd;border-radius:9px;padding:16px;margin-top:20px;">
<h3 style="color:#4338ca;margin:0 0 8px;font-size:17px;">🎯 Key Takeaway</h3>
<p style="font-size:14px;line-height:1.85;margin:0 0 10px;">Apache Spark provides a scalable way to process data by distributing computation across available resources. Its APIs let developers describe computations, while its execution engine plans and coordinates the work.</p>
