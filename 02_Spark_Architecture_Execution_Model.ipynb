<div style="font-size:16px;font-weight:750;color:#1d4ed8;margin-bottom:7px;">🎯 First, understand the two deployment modes</div>
<p style="font-size:13px;line-height:1.7;margin:0;color:#334155;">Spark deployment mode determines <strong>where the Driver process runs</strong>. It does not determine whether the application uses distributed cluster execution.</p>
<div style="display:flex;flex-wrap:wrap;gap:12px;margin-top:12px;">

<div style="font-size:14px;font-weight:750;color:#1d4ed8;margin-bottom:6px;">🖥️ Client Mode</div>
<div style="font-size:12px;color:#475569;line-height:1.7;">The Driver runs in the client process that submits the application. The client must remain available while the Driver is running.</div>
<div style="margin-top:9px;"><span style="display:inline-block;background:#dbeafe;color:#1d4ed8;border-radius:5px;padding:4px 7px;font-family:Consolas,monospace;font-size:12px;">--deploy-mode client</span></div>
</div>
<div style="box-sizing:border-box;flex:1 1 250px;background:#ffffff;border:1px solid #c4b5fd;border-radius:9px;padding:12px;">
<div style="font-size:14px;font-weight:750;color:#6d28d9;margin-bottom:6px;">☁️ Cluster Mode</div>
<div style="font-size:12px;color:#475569;line-height:1.7;">The Driver runs inside the cluster, managed by the cluster manager. The submitting client can disconnect after submission, subject to the deployment environment.</div>
<div style="margin-top:9px;"><span style="display:inline-block;background:#ede9fe;color:#6d28d9;border-radius:5px;padding:4px 7px;font-family:Consolas,monospace;font-size:12px;">--deploy-mode cluster</span></div>

<div style="flex:1;font-size:13px;line-height:1.7;color:#334155;"><strong>Initialize the Driver.</strong> The Driver establishes the Spark application context and coordinates work. It converts actions into jobs and organizes work into stages and tasks.</div>

<div style="display:flex;gap:10px;margin-bottom:12px;align-items:flex-start;">
<div style="flex:0 0 28px;height:28px;background:#d1fae5;color:#065f46;border-radius:8px;text-align:center;line-height:28px;font-weight:750;">4</div>
<div style="flex:1;font-size:13px;line-height:1.7;color:#334155;"><strong>Schedule tasks.</strong> The Driver's scheduling components assign tasks to executors based on available resources, dependencies and execution requirements.</div>
</div>
<div style="display:flex;gap:10px;margin-bottom:12px;align-items:flex-start;">
<div style="flex:0 0 28px;height:28px;background:#cffafe;color:#155e75;border-radius:8px;text-align:center;line-height:28px;font-weight:750;">5</div>
<div style="flex:1;font-size:13px;line-height:1.7;color:#334155;"><strong>Execute distributed work.</strong> Executors process their assigned partition-level tasks. Shuffle operations may move data between executors or worker nodes when required by the execution plan.</div>
</div>
<div style="display:flex;gap:10px;align-items:flex-start;">
<div style="flex:0 0 28px;height:28px;background:#fce7f3;color:#9d174d;border-radius:8px;text-align:center;line-height:28px;font-weight:750;">6</div>
<div style="flex:1;font-size:13px;line-height:1.7;color:#334155;"><strong>Complete the job.</strong> Task status and results are reported to the Driver. The Driver coordinates completion, returns results to the caller when applicable, and shuts down application resources when the application ends.</div>

<pre style="margin:0;white-space:pre-wrap;overflow-wrap:anywhere;font-family:Consolas,'Courier New',monospace;font-size:13px;line-height:1.8;"><code><span style="color:#6a9955;"># Driver runs in the submitting client process</span>
<span style="color:#dcdcaa;">spark-submit</span> <span style="color:#ce9178;">--master</span> <span style="color:#ce9178;">yarn</span> <span style="color:#ce9178;">--deploy-mode</span> <span style="color:#4ec9b0;">client</span> <span style="color:#9cdcfe;">my_app.py</span></code></pre>

Deployment-mode support and defaults depend on the cluster manager and Spark setup.</div>

<div style="font-size:15px;font-weight:750;color:#155e75;margin-bottom:8px;">🧠 Key takeaways</div>
<ul style="margin:0;padding-left:20px;color:#334155;font-size:13px;line-height:1.9;">
<li><strong>Driver:</strong> coordinates the Spark application and schedules distributed work.</li>
<li><strong>Cluster manager:</strong> allocates resources and helps launch application processes.</li>
<li><strong>Executors:</strong> run tasks and manage application data and intermediate results.</li>
<li><strong>Worker nodes:</strong> machines that host executor processes in a cluster.</li>
<li><strong>Deployment mode:</strong> determines where the Driver runs — client or cluster mode.</li>

<div style="font-size:24px;font-weight:700;line-height:1.35;">2.3 &nbsp; Cluster Managers &amp; Resource Allocation</div>
<div style="font-size:15px;line-height:1.6;margin-top:7px;color:#edf1ff;">Understand who provides the infrastructure, who manages cluster resources, and how Spark applications are deployed.</div>
</div>
<div style="box-sizing:border-box;width:100%;max-width:100%;border:1px solid #d9e2f2;border-top:0;border-radius:0 0 9px 9px;padding:18px;background:#fff;">
<div style="box-sizing:border-box;width:100%;padding:15px;border-left:5px solid #2864e8;background:#f0f6ff;border-radius:7px;margin-bottom:16px;">
<div style="font-size:18px;font-weight:700;color:#2055c7;margin-bottom:8px;">1. Who Provides the Cluster?</div>
<p style="margin:0 0 10px;line-height:1.7;">A Spark application needs computing infrastructure consisting of machines with CPU, RAM, storage and network connectivity. These may be physical servers or virtual machines provided by an organization or cloud provider.</p>
<p style="margin:0;line-height:1.7;"><strong>Important distinction:</strong> Infrastructure provides computing capacity; the cluster-management system manages or coordinates how that capacity is allocated to Spark applications.</p>
</div>
<div style="box-sizing:border-box;width:100%;border:1px solid #dce5f1;border-radius:8px;padding:14px;margin-bottom:18px;">
<div style="font-size:16px;font-weight:700;text-align:center;color:#263d83;margin-bottom:12px;">Two Layers of a Spark Cluster</div>
<div style="box-sizing:border-box;width:100%;padding:14px;background:#eaf1ff;border:1px solid #c9d9ff;border-radius:7px;text-align:center;">
<div style="font-weight:700;font-size:16px;color:#244eaa;">Layer 1 — Infrastructure</div>
<div style="font-size:14px;line-height:1.6;margin-top:5px;">Physical or virtual machines providing CPU, RAM, storage and networking</div>
</div>
<div style="text-align:center;padding:8px;font-size:20px;color:#687da8;">↓</div>
<div style="box-sizing:border-box;width:100%;padding:14px;background:#e8f8ef;border:1px solid #b8e5cb;border-radius:7px;text-align:center;">
<div style="font-weight:700;font-size:16px;color:#087747;">Layer 2 — Cluster Management</div>
<div style="font-size:14px;line-height:1.6;margin-top:5px;">Spark Standalone, YARN or Kubernetes manages or coordinates resource allocation and workload placement.</div>
</div>
</div>
<div style="font-size:18px;font-weight:700;color:#6b35c7;margin-bottom:10px;">2. Three Cluster-Management Options</div>
<div style="box-sizing:border-box;width:100%;border:1px solid #e1d5fa;border-radius:8px;padding:13px;margin-bottom:10px;">
<div style="font-weight:700;font-size:16px;color:#6b35c7;margin-bottom:6px;">A. Spark Standalone</div>
<div style="line-height:1.7;font-size:14px;">Spark's built-in cluster manager. It manages resources on registered Spark Worker Nodes and supports launching Executors on those workers.</div>
<div style="margin-top:8px;padding:9px;background:#f6f1ff;border-radius:5px;font-family:Consolas,monospace;font-size:13px;overflow-wrap:anywhere;">--master spark://master-host:7077</div>
<div style="font-size:12px;color:#52647c;margin-top:6px;">Replace the example hostname with your configured Standalone master address.</div>
</div>
<div style="box-sizing:border-box;width:100%;border:1px solid #bce6f2;border-radius:8px;padding:13px;margin-bottom:10px;">
<div style="font-weight:700;font-size:16px;color:#087e9d;margin-bottom:6px;">B. YARN</div>
<div style="line-height:1.7;font-size:14px;">A resource-management system in the Hadoop ecosystem. YARN allocates containers on cluster nodes, while Spark's deployment mechanisms launch the Driver and Executors in the appropriate environment.</div>
<div style="margin-top:8px;padding:9px;background:#eafaff;border-radius:5px;font-family:Consolas,monospace;font-size:13px;">--master yarn</div>
</div>
<div style="box-sizing:border-box;width:100%;border:1px solid #b9e7cf;border-radius:8px;padding:13px;margin-bottom:18px;">
<div style="font-weight:700;font-size:16px;color:#087747;margin-bottom:6px;">C. Kubernetes</div>
<div style="line-height:1.7;font-size:14px;">A container orchestration platform. Kubernetes schedules Pods onto cluster nodes. With Spark on Kubernetes, the Driver and Executors run in containers managed through Kubernetes Pods.</div>
<div style="margin-top:8px;padding:9px;background:#eaf9f0;border-radius:5px;font-family:Consolas,monospace;font-size:13px;overflow-wrap:anywhere;">--master k8s://https://k8s-api.example.com:6443</div>
<div style="font-size:12px;color:#52647c;margin-top:6px;">Example endpoint only; replace it with your configured Kubernetes API server address.</div>
</div>
<div style="box-sizing:border-box;width:100%;padding:15px;border-left:5px solid #df9411;background:#fff8e8;border-radius:7px;margin-bottom:16px;">
<div style="font-size:18px;font-weight:700;color:#9a6200;margin-bottom:8px;">3. What Does the Developer Configure?</div>
<p style="line-height:1.7;margin:0 0 10px;">The developer writes the Spark application and submits it using <code>spark-submit</code>. The command specifies the target cluster manager, deployment mode and applicable resource requests.</p>
<div style="box-sizing:border-box;width:100%;max-width:100%;background:#000000;border:1px solid #333333;border-radius:12px;padding:18px 16px;overflow-wrap:anywhere;">
  <pre style="margin:0;white-space:pre-wrap;overflow-wrap:anywhere;background:transparent !important;border:0 !important;box-shadow:none !important;border-radius:0 !important;padding:0 !important;font-family:Consolas,'Courier New',monospace;font-size:13px;line-height:1.8;color:#d4d4d4;"><code style="background:transparent !important;background-color:transparent !important;border:0 !important;box-shadow:none !important;border-radius:0 !important;padding:0 !important;color:inherit;"><span style="color:#dcdcaa;">spark-submit</span>
  <span style="color:#ce9178;">--master</span> <span style="color:#4ec9b0;">yarn</span>
  <span style="color:#ce9178;">--deploy-mode</span> <span style="color:#4ec9b0;">cluster</span>
  <span style="color:#ce9178;">--num-executors</span> <span style="color:#b5cea8;">6</span>
  <span style="color:#ce9178;">--executor-cores</span> <span style="color:#b5cea8;">4</span>
  <span style="color:#ce9178;">--executor-memory</span> <span style="color:#b5cea8;">8g</span>
  <span style="color:#dcdcaa;">my_app.py</span></code></pre>
</div>
<div style="font-size:14px;line-height:1.8;margin-top:10px;"><strong>--master yarn</strong> — selects YARN.<br/><strong>--deploy-mode cluster</strong> — runs the Driver inside the cluster.<br/><strong>--num-executors 6</strong> — requests six Executors.<br/><strong>--executor-cores 4</strong> — requests four CPU cores per Executor.<br/><strong>--executor-memory 8g</strong> — requests 8 GB JVM heap per Executor.<br/><strong>my_app.py</strong> — identifies the application code.</div>
<p style="line-height:1.7;margin:10px 0 0;"><strong>Note:</strong> These are resource requests, not guarantees. Actual allocation depends on resource availability and cluster policies.</p>
</div>
<div style="font-size:18px;font-weight:700;color:#087e76;margin-bottom:10px;">4. Who Does What?</div>
<div style="box-sizing:border-box;width:100%;max-width:100%;margin-bottom:16px;">
<table style="width:100%;border-collapse:collapse;font-size:14px;line-height:1.6;">

<div style="box-sizing:border-box;width:100%;padding:15px;border-left:5px solid #e35472;background:#fff0f3;border-radius:7px;margin-bottom:16px;">
<div style="font-size:18px;font-weight:700;color:#bf3453;margin-bottom:8px;">5. How Many Executors Run on Each Worker Node?</div>
<p style="line-height:1.7;margin:0 0 10px;">The standard YARN options above request a total number of Executors and resources per Executor. They do not directly specify a fixed number of Executors per Worker Node.</p>
<p style="line-height:1.7;margin:0 0 10px;">Actual placement depends on available CPU, memory, memory overhead, cluster-manager behavior and scheduling policies. A Worker Node may host multiple Executors if sufficient resources are available.</p>
<div style="box-sizing:border-box;width:100%;padding:12px;background:#fff;border:1px solid #f0c6d0;border-radius:6px;">
<strong>Example calculation</strong>
<p style="line-height:1.7;margin:7px 0;">6 Executors × 4 cores = <strong>24 requested Executor cores</strong><br/>6 Executors × 8 GB = <strong>48 GB requested JVM heap</strong></p>
<p style="font-size:13px;line-height:1.6;margin:0;color:#58657a;">Additional memory overhead is required. These totals do not determine the distribution of Executors across Worker Nodes.</p>

<div style="box-sizing:border-box;width:100%;padding:15px;border-left:5px solid #6478e7;background:#f0f2ff;border-radius:7px;margin-bottom:16px;">
<div style="font-size:18px;font-weight:700;color:#394eb8;margin-bottom:8px;">6. Deployment Mode Is Different from Cluster Manager</div>
<p style="line-height:1.7;margin:0 0 10px;"><code>--master</code> selects the cluster-management system or execution environment. <code>--deploy-mode</code> determines where the Driver runs, subject to the modes supported by that cluster manager and application type.</p>
<div style="box-sizing:border-box;width:100%;padding:10px;background:#fff;border:1px solid #d7ddff;border-radius:6px;margin-bottom:8px;"><strong style="color:#394eb8;">Client mode</strong><div style="font-size:14px;line-height:1.7;margin-top:4px;">The Driver runs in the submitting client process, typically on the submission machine.</div></div>
<div style="box-sizing:border-box;width:100%;padding:10px;background:#fff;border:1px solid #d7ddff;border-radius:6px;"><strong style="color:#394eb8;">Cluster mode</strong><div style="font-size:14px;line-height:1.7;margin-top:4px;">The Driver runs inside the cluster environment.</div></div>
</div>
<div style="box-sizing:border-box;width:100%;padding:15px;background:#eaf8ef;border:1px solid #bde6ca;border-radius:7px;">
<div style="font-size:18px;font-weight:700;color:#087747;margin-bottom:8px;">Key Takeaways</div>
<ul style="padding-left:20px;margin:0;line-height:1.8;">
<li>Infrastructure provides the machines; cluster management coordinates resource allocation.</li>
<li>Spark Standalone, YARN and Kubernetes use different resource-management and deployment mechanisms.</li>
<li>The developer requests resources; actual allocation and placement depend on the cluster infrastructure.</li>
<li>The standard YARN submission command does not fix the number of Executors per Worker Node.</li>
<li>The Driver coordinates the application; Executors execute its tasks.</li>
<li>Cluster-manager choice and Driver deployment mode are separate configuration decisions.</li>
</ul>

2.4.1 🧭 Two Connected Lifecycles: Infrastructure and Application
</h2>
<p style="box-sizing:border-box;font-family:'Segoe UI', Arial, sans-serif; font-size:15px; line-height:1.8; color:#334155; margin:0 0 16px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
A Spark application does not normally create an entire cluster from scratch. Infrastructure preparation and Spark application execution are two connected but distinct activities.
</p>
<div style="width:100%;box-sizing:border-box;border:1px solid #cbd5e1;border-radius:10px;padding:17px;margin:14px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<div style="box-sizing:border-box;font-size:12px;font-weight:700;letter-spacing:1.4px;color:#0875a8;;max-width:100%;min-width:0;overflow-wrap:anywhere">LIFECYCLE A</div>
<h3 style="box-sizing:border-box;font-size:19px;color:#174f83;margin:5px 0 8px;;max-width:100%;min-width:0">Infrastructure / Platform Setup</h3>
<p style="box-sizing:border-box;margin:0 0 12px;;max-width:100%;min-width:0;overflow-wrap:anywhere">The platform or DevOps administrator provisions physical or virtual machines, installs and configures the required software, configures the cluster manager, establishes networking and security, and starts the required services.</p>
<div style="width:100%;box-sizing:border-box;background:#eff6ff;border-radius:7px;padding:11px 13px;;max-width:100%;min-width:0;overflow-wrap:anywhere">Result: a cluster environment is ready to accept Spark applications.</div>
</div>
<div style="box-sizing:border-box;text-align:center;font-size:24px;color:#0875a8;margin:7px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">↓</div>
<div style="width:100%;box-sizing:border-box;border:1px solid #cbd5e1;border-radius:10px;padding:17px;margin:14px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<div style="box-sizing:border-box;font-size:12px;font-weight:700;letter-spacing:1.4px;color:#0875a8;;max-width:100%;min-width:0;overflow-wrap:anywhere">LIFECYCLE B</div>
<h3 style="box-sizing:border-box;font-size:19px;color:#174f83;margin:5px 0 8px;;max-width:100%;min-width:0">Spark Application Submission and Execution</h3>
<p style="box-sizing:border-box;margin:0 0 12px;;max-width:100%;min-width:0;overflow-wrap:anywhere">The developer writes Spark code, prepares the application, and submits it using <code style="background:#eef2f7;padding:2px 5px;border-radius:4px;">spark-submit</code> or another supported entry point. Spark's deployment and runtime mechanisms arrange the Driver and Executor resources according to the selected cluster manager and deployment mode.</p>
<div style="width:100%;box-sizing:border-box;background:#eff6ff;border-radius:7px;padding:11px 13px;;max-width:100%;min-width:0;overflow-wrap:anywhere">Result: the application executes its Spark jobs using allocated resources.</div>
</div>
<div style="box-sizing:border-box;background:#fff7ed;border:1px solid #fed7aa;border-left:5px solid #ea8a23;border-radius:8px;padding:13px 15px;margin:17px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<strong style="color:#9a4a0b;">Important distinction:</strong> In a typical enterprise setup, the cluster manager and its infrastructure services are already running before the developer submits an application. Selecting a cluster manager does not, by itself, provision the machines or install the platform.
</div>
<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.2 What Does the Platform / DevOps Administrator Configure?</h2>
<p>The exact work differs by platform, but the responsibilities usually include the following.</p>
<ul>
<li><strong>Infrastructure:</strong> Provision physical or virtual machines, or prepare cloud compute resources.</li>
<li><strong>Software:</strong> Install and configure Spark and any required supporting software.</li>
<li><strong>Cluster manager:</strong> Configure and start Standalone services, Hadoop YARN services, or a Kubernetes environment.</li>
<li><strong>Networking:</strong> Ensure the Driver, Executors, cluster manager, and required data services can communicate.</li>
<li><strong>Security:</strong> Configure identities, permissions, authentication, and access to data and services.</li>
<li><strong>Capacity:</strong> Establish available CPU, memory, storage, and operational limits.</li>
<li><strong>Application access:</strong> Ensure the developer can submit jobs and access the required dependencies and data.</li>
</ul>
<p>In a small learning environment, one person may perform both platform administration and Spark development. In an enterprise, these responsibilities are often separated.</p>
<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.3 Common Submission Concepts</h2>

<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">The Driver runs in the submitting client process. The client machine or gateway must remain available for the Driver while the application runs.</p>

<div style="width:100%;box-sizing:border-box;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:14px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 8px;;max-width:100%;min-width:0">Cluster deployment mode</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">The Driver runs inside the cluster's execution environment. The exact process or container that hosts it depends on the cluster manager and the deployment mode.</p>
</div>
<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.4 Scenario A — Spark Standalone</h2>
<p>Spark Standalone is Spark's own cluster manager. Its infrastructure includes a central <strong>Master</strong> process and one or more <strong>Worker</strong> processes.</p>
<div style="width:100%;box-sizing:border-box;background:#f8fafc;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:13px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">How the components fit together</h3>
<ul style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0">
<li><strong>Master:</strong> the central coordinating process of the Spark Standalone cluster manager. It tracks registered Workers and helps allocate application resources.</li>
<li><strong>Worker:</strong> a Standalone service running on a machine that manages the resources available there and launches application processes as directed.</li>
<li><strong>Driver:</strong> coordinates the Spark application. Its location depends on the deployment mode.</li>
<li><strong>Executors:</strong> application-specific processes that run tasks and may cache data.</li>
</ul>
<p style="box-sizing:border-box;margin:10px 0 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">For the general meaning of a Driver, Executor, task, and job, refer to Section 2.1 and the Spark application lifecycle in Section 2.2.</p>
</div>

<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.5 Scenario B — Hadoop YARN</h2>
<p><strong>YARN</strong> stands for <em>Yet Another Resource Negotiator</em>. It is a cluster resource-management system commonly used in Hadoop environments. Spark can use YARN to obtain resources for its applications.</p>
<div style="width:100%;box-sizing:border-box;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:13px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 8px;;max-width:100%;min-width:0">YARN components</h3>
<ul>
<li><strong>ResourceManager (RM):</strong> the cluster-wide YARN service responsible for resource scheduling and allocation across applications.</li>
<li><strong>NodeManager (NM):</strong> a service on each participating machine that manages local containers and reports resource information to the ResourceManager.</li>
<li><strong>ApplicationMaster (AM):</strong> an application-specific YARN component that negotiates resources with the ResourceManager and coordinates the application's YARN-side execution.</li>
<li><strong>YARN container:</strong> a bundle of allocated resources, such as memory and CPU, on a particular node in which an application process can run. It is not the same concept as a Docker container.</li>
</ul>
</div>

</div>
<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.6 Scenario C — Kubernetes</h2>
<p>Kubernetes manages containerized workloads. Spark can run its Driver and Executors as Kubernetes Pods, with each Pod scheduled onto a cluster node.</p>
<div style="width:100%;box-sizing:border-box;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:13px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 8px;;max-width:100%;min-width:0">Kubernetes terminology</h3>
<ul>
<li><strong>Control Plane:</strong> the group of Kubernetes components that maintains the desired cluster state, accepts API requests, makes scheduling decisions, and coordinates cluster management.</li>
<li><strong>Node:</strong> a machine in the Kubernetes cluster, physical or virtual, that can run Pods.</li>
<li><strong>Pod:</strong> the smallest deployable unit in Kubernetes. It contains one or more closely related containers that share certain resources, such as networking. A Spark Driver or Executor typically runs in its own Pod.</li>
<li><strong>Container:</strong> a packaged, isolated process environment created from an image. A Pod runs its container or containers on a node.</li>
<li><strong>Scheduler:</strong> a Kubernetes Control Plane component that selects a suitable node for a Pod based on resource requests, constraints, and availability.</li>
</ul>
</div>

<h3 style="box-sizing:border-box;font-size:18px;color:#174f83;margin:18px 0 8px;;max-width:100%;min-width:0">What happens?</h3>
<ol>
<li>The submission process requests that Spark run on the configured Kubernetes cluster.</li>
<li>Kubernetes schedules the Driver Pod onto a suitable node.</li>
<li>The Spark Driver initializes the application and requests Executor resources.</li>
<li>Kubernetes schedules Executor Pods onto nodes that satisfy the resource and placement requirements.</li>
<li>The Driver coordinates Spark tasks, and the Executors perform the distributed work.</li>
</ol>

<strong style="color:#174f83;">Key distinction:</strong> A Kubernetes Pod is a Kubernetes deployment unit; a container is the process environment inside a Pod. A YARN container is instead a YARN resource-allocation concept. The two systems use the word “container” in different technical senses.

<div style="box-sizing:border-box;font-size:13px;color:#526577;margin-top:4px;;max-width:100%;min-width:0;overflow-wrap:anywhere">The Driver initializes and coordinates the Spark application. SparkContext provides the connection to core Spark execution services.</div>
</div>
<div style="box-sizing:border-box;text-align:center;font-size:22px;color:#0875a8;line-height:1.4;margin:4px 0;;max-width:100%;min-width:0;overflow-wrap:anywhere">↓</div>
<div style="width:100%;box-sizing:border-box;background:#ffffff;border:1px solid #b9d9f5;border-radius:8px;padding:13px;text-align:center;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<div style="box-sizing:border-box;font-size:18px;font-weight:700;color:#174f83;;max-width:100%;min-width:0;overflow-wrap:anywhere">Executors</div>
<div style="box-sizing:border-box;font-size:13px;color:#526577;margin-top:4px;;max-width:100%;min-width:0;overflow-wrap:anywhere">The Executors execute the tasks assigned to them by the Driver and return status and results as required.</div>
</div>
</div>
<p>This is a conceptual model, not a strict four-step sequence or four independent services. Resource allocation, process launch, Driver initialization, and Executor startup differ by platform and deployment mode and can overlap.</p>

<h2 style="box-sizing:border-box;font-size:24px;font-weight:400;color:#0875a8;border-bottom:2px solid #b9d9f5;padding-bottom:9px;margin:28px 0 14px;;max-width:100%;min-width:0">2.4.11 Questions &amp; Answers</h2>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q1. Does the developer initialize SparkContext separately for each cluster manager, and what does SparkSession.builder.getOrCreate() do?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">No. The developer typically creates or obtains a SparkSession using SparkSession.builder.getOrCreate(). This method returns an existing suitable SparkSession in the current application context when available; otherwise, it creates a session. During new-session setup, Spark initializes or reuses the underlying Driver-side SparkContext as needed. A suitable existing context is one available and compatible within that application's Driver process—not a context searched for across the cluster. The same programming pattern works with Standalone, YARN, and Kubernetes; deployment settings determine where the Driver runs and how it connects to the selected environment.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q2. In YARN client mode, is the ApplicationMaster absent?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">No. YARN still has an ApplicationMaster for the application. The distinction is that in client mode, the Spark Driver runs in the submitting client process; in cluster mode, the Driver runs inside the ApplicationMaster process.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q3. Does “Spark application” mean the code the developer submits?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">The submitted source file or packaged program is the application's code. “Spark application” commonly refers to that program together with its running application instance, including its Driver and Executors.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q4. How can the cluster manager allocate resources before SparkContext exists, and can a Spark application execute without SparkContext?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">A Spark application can be submitted before its Driver-side SparkContext exists, but a normal Spark application cannot execute Spark jobs without a Driver-side SparkContext. Submission provides the application and deployment/resource configuration, allowing the cluster manager to arrange the initial Driver launch in cluster mode (for example, a YARN ApplicationMaster container or Kubernetes Driver Pod). In client mode, the Driver runs in the submitting client process. Once the Driver process starts, the Spark runtime initializes or obtains SparkSession and SparkContext. The Driver-side context then participates in requesting Executor resources; Executors start and run tasks. Thus, initial Driver launch and subsequent Executor allocation are distinct phases. A script may run ordinary code or even exit without ever creating a SparkContext, but it cannot perform normal distributed Spark computation without one.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q5. Is a cluster made of physical or virtual machines?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">It can be either or a combination. A cluster is a set of connected machines or nodes coordinated to provide compute resources. In cloud environments, virtual machines are common. Kubernetes nodes can also be virtual or physical machines.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q6. What is a Kubernetes Pod, and how does it work?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">A Pod is Kubernetes' smallest deployable unit. It contains one or more containers that share certain resources. Kubernetes schedules the Pod onto a node, and the container runtime runs its containers. In a typical Spark-on-Kubernetes deployment, the Driver runs in a Driver Pod and Executors run in Executor Pods. The Driver coordinates Spark tasks executed by the Executors.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q7. Does the YARN ResourceManager coordinate with the ApplicationMaster? Is its role the same in client and cluster mode?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;;max-width:100%;min-width:0;overflow-wrap:anywhere">The ResourceManager is the cluster-wide resource scheduler. The ApplicationMaster requests and negotiates resources for its application with the ResourceManager and coordinates the application's resource use.</p>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">The ResourceManager's cluster-wide role remains broadly the same in client and cluster deployment modes. The main difference is the Driver's location: in cluster mode, the Driver is hosted in the ApplicationMaster process; in client mode, the Driver runs in the client process while the ApplicationMaster still exists for YARN resource coordination.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q8. Is a YARN container the same as a Docker container or a Kubernetes container?</h3>
<p style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0;overflow-wrap:anywhere">No. A YARN container represents an allocation of resources on a node for running an application process. A Kubernetes container is a running process environment created from an image and managed inside a Pod. Although the same word is used, the concepts are not interchangeable.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;max-width:100%;min-width:0">Q9. What are YARN containers, Kubernetes containers, and Docker containers? How are they different?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>YARN container:</strong> A YARN-managed allocation of resources—such as memory and CPU—on a particular node, together with the launch context used to start an application process. It is primarily a resource-allocation and process-launch concept, not inherently a Docker-style container environment. A NodeManager launches and monitors the process in that allocation; for Spark, that process may be an Executor or, in YARN cluster mode, the application-specific ApplicationMaster process that hosts the Driver.</p>
<p style="box-sizing:border-box;margin:0 0 8px;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Docker container:</strong> A running, isolated process environment created from a Docker image. The image packages the application and its required filesystem layers and configuration; the container runtime starts the container using operating-system isolation mechanisms. Docker is a container platform/tooling ecosystem, and a container is not itself a physical machine or a cluster manager.</p>
<p style="box-sizing:border-box;margin:0 0 8px;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Kubernetes container:</strong> A container runtime's running container managed as part of a Kubernetes Pod. Kubernetes schedules the Pod—not each container independently—onto a node. A Pod contains one or more containers that can share selected resources. Kubelet on the assigned node works with the container runtime to ensure the Pod's containers are running.</p>
<p style="box-sizing:border-box;margin:0;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Key distinction:</strong> A traditional YARN container describes a node-level resource allocation and process launch; Docker and Kubernetes containers describe image-based, operating-system-isolated process environments. Kubernetes uses container runtimes to run containers inside Pods. YARN can also be configured to launch Docker containers, so the concepts can overlap in a configured system, but the terms do not mean the same thing by default.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;;max-width:100%;min-width:0">Q10. What is the generalized working model of Spark across Standalone, YARN, and Kubernetes?</h3>
<ol style="box-sizing:border-box;margin:0;;max-width:100%;min-width:0">
<li>The platform administrator prepares the infrastructure and starts the cluster-management services.</li>
<li>The developer prepares and submits Spark application code.</li>
<li>The deployment integration arranges for the Driver to run in the selected location.</li>
<li>The Spark runtime initializes the Driver-side SparkSession and SparkContext as needed.</li>
<li>The Driver plans application work and requests or uses Executor resources.</li>
<li>The platform launches Executors using its own mechanisms: Standalone Workers, YARN-managed containers, or Kubernetes Pods.</li>
<li>The Driver schedules tasks; Executors process partitions and return status or results.</li>
<li>The application finishes, and its processes and allocated resources are cleaned up.</li>
</ol>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q11. Who decides whether an application uses client or cluster deployment mode? Is the mode fixed when the cluster is installed?</h3>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">The mode is not universally fixed at cluster-installation time. The platform/DevOps team prepares the infrastructure and configures the supported execution options. Platform governance, security, administrators, or managed-service controls may also define which modes are allowed. For each application, the developer, job configuration, submission tool, or orchestrator requests a supported mode. Some organizations establish the policy during initial setup; others introduce or update it later.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q12. If an organization requires cluster mode, can a developer switch to client mode to run the same code?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">Only if the platform permits client mode and the applicable policy allows it. If a submission gateway, job platform, cluster policy, or managed service enforces cluster mode, a client-mode request can be rejected or blocked. Changing the flag or writing different Spark application code does not override an enforced policy. The developer would need an authorized policy exception or an approved execution path.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">A policy may be only a documented convention, or it may be technically enforced. Those are different situations: a convention might not stop a job technically, but running against it would still violate organizational rules.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">

<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">For a compatible Spark submission, client or cluster mode is commonly requested through the submission configuration, such as <code style="background:transparent !important;color:inherit;padding:0;border:0;box-shadow:none">--deploy-mode</code> in <code style="background:transparent !important;color:inherit;padding:0;border:0;box-shadow:none">spark-submit</code>. It is a deployment choice, not normally a change to the application's business logic.</p>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q14. Why would a team prefer cluster mode over client mode, or client mode over cluster mode?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Cluster mode</strong> is often chosen for scheduled or production jobs because the Driver runs in the cluster-managed environment and the submitting client generally does not need to remain connected for the entire application lifetime.</p>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Client mode</strong> is often convenient for interactive development, debugging, and observing output directly in a client or gateway process. That client environment must remain available while it hosts the Driver, and network connectivity between the Driver and cluster must work.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">These are common patterns, not absolute rules. The decision usually reflects operational and technical requirements—such as where the Driver should run, job lifetime, monitoring, networking, environment dependencies, and platform policy—not simply a business requirement in isolation. Client mode can be used in production when the platform design supports it.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q15. Does a scheduled Azure Data Factory (ADF) job automatically mean Spark cluster mode?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">No. ADF is an orchestration service: it can schedule and trigger a pipeline, but scheduling does not itself determine where a Spark Driver runs. The target compute service and its job configuration determine the execution model.</p>
<ul style="box-sizing:border-box;margin:0 0 8px;min-width:0">
<li><strong>ADF triggering an Azure Databricks job or notebook:</strong> Databricks job and compute settings govern execution; the user does not normally choose the low-level deploy mode by editing the notebook's Spark business logic.</li>
<li><strong>ADF submitting a Spark application to HDInsight:</strong> the supported Spark submission options and cluster configuration govern how the application is launched.</li>
</ul>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">Therefore, distinguish the scheduler (when and in what workflow the job runs) from deployment mode (where the Driver runs). Check the target service's configuration rather than inferring the mode from the fact that ADF runs it on a schedule.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q16. Does installing a cluster automatically create a policy that requires cluster mode?</h3>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">No. Cluster installation prepares the environment and its services; a mandatory deployment-mode policy is a separate configuration or governance decision. A platform may support both modes without imposing a universal policy, or an organization may restrict the permitted mode through job templates, submission gateways, access controls, managed-service policies, or operating procedures. The exact mechanism depends on the platform.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q17. In Kubernetes, what does “desired cluster state” mean?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">The desired state is the condition you declare that Kubernetes should maintain. For example, a configuration may request three replicas of a workload, a particular container image, and specified resource requirements.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">Kubernetes compares the desired state with the actual state. If one replica stops running, a controller can detect the difference and arrange for a replacement. This ongoing process is called <strong>reconciliation</strong>. The Control Plane coordinates this work; it does not mean that every Control Plane component personally performs each repair.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q18. What is the difference between the Kubernetes Control Plane and a Pod?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">They operate at different layers:</p>
<ul style="box-sizing:border-box;margin:0 0 8px;min-width:0">
<li><strong>Control Plane:</strong> manages and coordinates the cluster. Its components process API requests, maintain the desired state, and make or coordinate scheduling decisions.</li>
<li><strong>Pod:</strong> the smallest Kubernetes deployable unit. It hosts one or more containers that run workload processes, such as a Spark Driver or Executor.</li>
</ul>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">In short, the Control Plane manages the cluster; a Pod is a unit in which an application component runs. The scheduler is a Control Plane component that selects a suitable node for a Pod that needs placement.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q19. Does “three replicas” mean the developer submitted three Spark applications?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>No, not necessarily.</strong> “Three replicas” means three copies of the workload described by the relevant Kubernetes controller or configuration. It does not, by itself, mean three separate Spark application submissions.</p>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">For a typical Spark application on Kubernetes, one Driver coordinates the application and multiple Executor Pods perform distributed work. For example, one Spark application may have one Driver Pod and three Executor Pods.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">Do not assume that a generic Kubernetes Deployment with three replicas is the normal way to manage Spark Executors. Spark's Kubernetes integration normally creates and manages Executor Pods according to Spark's executor configuration and lifecycle. Three separately submitted Spark application executions would typically have separate Drivers.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q20. Are the mappings between Pods, Driver/Executor processes, and containers one-to-one or one-to-many?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Pod to Spark Driver/Executor process:</strong> in the standard Spark-on-Kubernetes deployment model, the mapping is typically one-to-one: a Driver process runs in a Driver Pod, and each Executor process runs in its own Executor Pod. One Spark application commonly has one active Driver and multiple Executor Pods. Kubernetes itself is more general, so distinguish this common Spark deployment pattern from every arrangement Kubernetes technically permits.</p>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Pod to container:</strong> one-to-one and one-to-many are both supported. A Pod must contain at least one container and can contain multiple closely related containers. For example, an Executor Pod may contain the Executor container and an optional supporting sidecar container for logging or monitoring.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">Containers in the same Pod share certain Pod-level resources, including the network namespace and IP address, while remaining separate container environments.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q21. How does a Spark application execute on Spark Standalone? Describe the components and communication flow in client and cluster deploy modes.</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">Spark Standalone has a <strong>Master</strong> service and <strong>Worker</strong> services for cluster resource management. The <strong>Driver</strong> runs the application’s main logic, creates the SparkContext/SparkSession, plans Jobs and Stages when actions trigger work, and schedules individual Spark Tasks. The Master does not plan queries or schedule individual Spark Tasks. Unlike YARN, Standalone has <strong>no separate NodeManager service</strong>: the Standalone Worker daemon is the node-level Spark service that launches and monitors application processes, including Executors and (in cluster deploy mode) the Driver.</p>
<h4 style="color:#174f83;font-size:16px;margin:14px 0 8px">A. Standalone client mode — Driver runs in the submitting client process</h4>
<div style="width:100%;box-sizing:border-box;background:#eff6ff;border:1px solid #bfdbfe;border-radius:8px;padding:12px;margin:10px 0;overflow-wrap:anywhere;min-width:0;max-width:100%">
<div style="font-weight:700;color:#174f83;margin-bottom:7px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">1. Identify the components first</div>
<div style="display:flex;flex-wrap:wrap;gap:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">
<div style="flex:1 1 240px;background:#ffffff;border:1px solid #93c5fd;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component A — Submission host</strong><br/><strong>spark-submit + Driver process</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">spark-submit identifies the application resource and starts the Driver in this client-side process. The Driver loads/runs the application's entry point (top-level Python script or JVM main method), initializes Spark, and coordinates execution. This host may or may not also be a Worker node; the Master does not launch this Driver.</div></div>
<div style="flex:1 1 240px;background:#dbeafe;border:1px solid #93c5fd;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component B — Standalone Master service/process</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Tracks Worker resource advertisements/heartbeats, accepts cluster-mode submission requests, selects available capacity, and coordinates Driver/Executor launches. It does not build Spark query plans or dispatch individual Spark Tasks.</div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component C — Worker node A</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Standalone Worker daemon receives launch instructions and launches/monitors an Executor process on this node.</div><div style="background:#ffffff;border:1px solid #86efac;border-radius:5px;padding:7px;margin-top:6px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Executor process A</strong><br/>Runs Spark Tasks on partitions</div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component D — Worker node B</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Standalone Worker daemon receives launch instructions and launches/monitors an Executor process on this node.</div><div style="background:#ffffff;border:1px solid #86efac;border-radius:5px;padding:7px;margin-top:6px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Executor process B</strong><br/>Runs Spark Tasks on partitions</div></div>
</div>
<div style="font-weight:700;color:#174f83;margin:12px 0 7px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">2. Follow the communications in sequence</div>
<div style="background:#ffffff;border:1px solid #93c5fd;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>① Client process / Driver → Standalone Master:</strong> after the Driver starts and initializes Spark, its Standalone scheduler backend registers the application with the configured Master URL (for example, <code>spark://Master-Host:7077</code>) and requests Executor resources. The Driver is already running on the submitting host; there is no cluster-side Driver allocation or launch in client mode.</div>
<div style="text-align:center;color:#0875a8;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↑ Worker resource reports / heartbeats; ↓ launch instructions</div>
<div style="background:#ffffff;border:1px solid #93c5fd;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>② Workers ↔ Master; Master → selected Worker services:</strong> Worker daemons register with the Master and report available resources. The Master uses that information and the application’s resource configuration to choose eligible Workers and instruct them to launch Executor processes. Each selected Worker daemon performs the local process launch and monitors that Executor.</div>
<div style="text-align:center;color:#0875a8;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↕ Direct network communication</div>
<div style="background:#ffffff;border:1px solid #93c5fd;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>③ Executors ↔ Driver:</strong> each Executor process connects to and registers with the Driver directly. Executors do not send the query to the Master for planning. The Driver learns which Executors are available and their task capacity; Executors send status and results back to the Driver.</div>
<div style="text-align:center;color:#0875a8;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ User application code invokes an action in the Driver</div>
<div style="background:#fef3c7;border:1px solid #fcd34d;border-radius:7px;padding:10px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>④ Application code / action → Driver planning → TaskScheduler → Executors</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">The application code calls Spark APIs in the Driver process. When an action triggers computation, Driver-side components plan Jobs and Stages; the TaskScheduler sends individual Tasks directly to registered Executors. Executors process partitions and exchange shuffle data when required. The Master is not in the per-query planning or per-task dispatch path.</div></div>
<div style="background:#f8fafc;border:1px solid #cbd5e1;border-radius:6px;padding:9px;margin-top:8px;font-size:13px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Key point:</strong> Client mode has no cluster-side Driver launch. The Driver runs the application entry point on the submitting host and requests cluster resources through Spark's Standalone integration; the Master selects Workers and coordinates Executor launches, while Worker daemons launch Executor processes. The client host must be network-reachable from Executors and can, but need not, also be a Worker.</div>
</div>
<h4 style="color:#174f83;font-size:16px;margin:14px 0 8px">B. Standalone cluster mode — Master coordinates Driver launch on a Worker</h4>
<div style="width:100%;box-sizing:border-box;background:#f0fdf4;border:1px solid #86efac;border-radius:8px;padding:12px;margin:10px 0;overflow-wrap:anywhere;min-width:0;max-width:100%">
<div style="font-weight:700;color:#166534;margin-bottom:7px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">1. Identify the components first</div>
<div style="display:flex;flex-wrap:wrap;gap:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">
<div style="flex:1 1 240px;background:#ffffff;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component A — Submission host</strong><br/><strong>spark-submit client process</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Submits the application, cluster deploy mode, and configuration to the Standalone Master. It is not the process that hosts the application Driver.</div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component B — Standalone Master service/process</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Accepts the cluster-mode submission, selects a suitable Worker for the Driver, and coordinates Driver and Executor process launches based on cluster resources.</div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component C — Worker node A</strong><br/><strong>Standalone Worker daemon</strong><div style="background:#fff7ed;border:1px solid #fdba74;border-radius:5px;padding:8px;margin-top:6px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Driver process</strong><div style="font-size:13px;margin-top:3px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">The Worker daemon launches this process on its node. After launch, the Driver process initializes Spark and runs application/query logic.</div></div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component D — Worker node B</strong><br/><strong>Standalone Worker daemon</strong><div style="background:#ffffff;border:1px solid #86efac;border-radius:5px;padding:7px;margin-top:6px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Executor process B</strong><br/>Runs Spark Tasks on partitions</div></div>
<div style="flex:1 1 240px;background:#dcfce7;border:1px solid #86efac;border-radius:7px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Component E — Worker node C</strong><br/><strong>Standalone Worker daemon</strong><div style="background:#ffffff;border:1px solid #86efac;border-radius:5px;padding:7px;margin-top:6px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Executor process C</strong><br/>Runs Spark Tasks on partitions</div></div>
</div>
<div style="font-weight:700;color:#166534;margin:12px 0 7px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">2. Follow the communications in sequence</div>
<div style="background:#ffffff;border:1px solid #86efac;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>① Submission client → Standalone Master:</strong> sends the cluster-mode submission request, application resource reference, and deployment/resource configuration. The Master coordinates where the Driver should run; it does not itself execute the Driver application or act as the application's query engine.</div>
<div style="text-align:center;color:#15803d;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ launch coordination</div>
<div style="background:#ffffff;border:1px solid #86efac;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>② Master → selected Worker daemon → Driver process:</strong> the Master selects a suitable Worker and coordinates launch. The Worker daemon launches and monitors the Driver locally. The application resource must be available to the launched process: for JARs submitted with Spark's Standalone spark-submit mechanism, Spark distributes the application JAR to Worker nodes; other resources/dependencies must be accessible through the applicable staging/URI mechanism. The Master need not hold the only copy of the JAR. This Worker is analogous in one limited role to a node-level process launcher such as YARN's NodeManager, but it is not a NodeManager.</div>
<div style="text-align:center;color:#15803d;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ after the Driver process has started</div>
<div style="background:#ffffff;border:1px solid #86efac;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>③ Driver process initializes Spark:</strong> the launched Driver starts the application’s Spark runtime, creates/initializes SparkContext and the relevant Driver-side scheduling components, then runs the application’s main logic. Driver launch and Spark initialization are distinct steps.</div>
<div style="text-align:center;color:#15803d;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↕ resource management / direct Driver communication</div>
<div style="background:#ffffff;border:1px solid #86efac;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>④ Driver ↔ Master; Master → Worker daemons:</strong> the Driver’s Standalone scheduler backend requests Executor resources from the Master. The Master makes cluster resource decisions and instructs selected Worker daemons to launch Executor processes; each Worker daemon launches and monitors the process on its own node. Resource provisioning can overlap with application/query planning.</div>
<div style="text-align:center;color:#15803d;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↕ direct network communication</div>
<div style="background:#ffffff;border:1px solid #86efac;border-radius:6px;padding:9px;margin:6px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>⑤ Executor processes ↔ Driver:</strong> Executors connect to and register with the Driver directly. They report status and results to the Driver. The Master does not relay the application’s query to the Driver and is not in the per-task scheduling path.</div>
<div style="text-align:center;color:#15803d;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ application code invokes an action in the Driver</div>
<div style="background:#fef3c7;border:1px solid #fcd34d;border-radius:7px;padding:10px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>⑥ Application code / action → Driver planning → TaskScheduler → Executors</strong><div style="font-size:13px;margin-top:4px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">The application’s query/API calls run in the Driver process. When an action triggers computation, the Driver plans Jobs and Stages and the TaskScheduler dispatches Tasks directly to registered Executors. Executors compute on partitions and exchange shuffle data when needed.</div></div>
</div>
<div style="background:#f8fafc;border:1px solid #cbd5e1;border-radius:7px;padding:10px;margin:8px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Direct answer: how does the application/query reach the Driver?</strong> The application code is executed by the Driver process: in client mode it is already running in the submitting client process; in cluster mode the submission is used to launch the Driver on a Worker, and the Driver then runs the application code. A query is not sent by an Executor to the Driver, nor does the Master forward each query for planning. The application invokes Spark APIs in the Driver, and Driver-side components plan and schedule work. The Driver then communicates directly with registered Executors to dispatch Tasks and receive task status/results.</div>
<p style="margin:0 0 8px"><strong>Lifecycle summary:</strong> <em>Client mode:</em> client starts Driver → Driver initializes Spark and registers/requests resources from Master → Master selects Workers and coordinates Executor launches → Worker daemons launch Executors → Executors register with Driver → Driver plans Jobs/Stages and schedules Tasks → Executors execute Tasks. <em>Cluster mode:</em> client submits to Master → Master selects Worker → Worker daemon launches Driver → Driver initializes Spark and requests Executors → Master coordinates Worker-side Executor launches → Executors register with Driver → Driver plans and schedules work → Executors execute Tasks.</p>
<p style="margin:0"><strong>Remember the three layers:</strong> <em>Master/Worker resource management</em> decides where application processes can run; <em>Driver-side Spark planning</em> decides what computation is required; <em>TaskScheduler</em> assigns individual Spark Tasks to registered Executors. The Standalone Worker daemon is the local launcher/monitor; there is no separate NodeManager in Standalone.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q22. What is an application entry point? Show a Spark example.</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>Entry point</strong> means the place where the application begins running its own code. In PySpark, that is commonly the top-level statements in the submitted Python script; in Scala/Java, it is commonly a <code>main()</code> method.</p>
<pre style="white-space:pre-wrap;overflow-wrap:anywhere;background:#0b1220;color:#e5e7eb;border-radius:8px;padding:12px;font-size:13px;line-height:1.6"><code># sales_app.py — simplified example
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SalesCount").getOrCreate()
df = spark.read.parquet("sales.parquet")
result = df.groupBy("region").count()
result.show()  # action triggers distributed work
spark.stop()</code></pre>
<p style="margin:0 0 8px;line-height:1.65">The Driver process runs this script and coordinates Spark. The Executors run the distributed Tasks that read/process partitions. The entry point is not the same as the distributed data-processing work.</p>
<p style="margin:0;line-height:1.65"><strong>Scala/Java contrast:</strong> the corresponding entry point is typically <code>def main(args: Array[String]): Unit</code> in Scala or <code>public static void main(String[] args)</code> in Java.</p>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q23. In Standalone cluster mode, how does the application resource reach the Driver on a Worker?</h3>
<p style="margin:0 0 8px;line-height:1.65">The submission client sends the cluster-mode submission request and launch information to the Standalone Master. The Master selects a Worker and coordinates the Driver launch; the selected Worker daemon starts the Driver process with the application resource and configuration available to it.</p>
<p style="margin:0 0 8px;line-height:1.65">For a JAR submitted through Spark's Standalone <code>spark-submit</code> mechanism, Spark documentation states that the application JAR is distributed to Worker nodes. Additional JAR dependencies can be supplied using <code>--jars</code>. For Python files, external resources, and other dependencies, their accessibility/staging depends on the submission mechanism and resource URI; do not assume the Master stores the only copy and forwards it.</p>
<div style="background:#eff6ff;border:1px solid #bfdbfe;border-radius:6px;padding:10px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Sequence:</strong> submission client provides the resource reference → Master selects Worker → Worker daemon launches Driver → Driver loads the application entry point. Process launch and resource staging/availability are related but distinct operations.</div>
<p style="margin:8px 0 0;line-height:1.65">Official reference: <a href="https://spark.apache.org/docs/latest/spark-standalone.html" rel="noopener noreferrer" target="_blank">Spark Standalone Mode — Launching Spark Applications</a>.</p>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q24. What is the Standalone Worker daemon, and is there one daemon for the whole cluster?</h3>
<p style="margin:0 0 8px;line-height:1.65">A <strong>daemon</strong> is a long-running background process. A Standalone Worker daemon runs on each configured Worker host/instance, registers with the Master, reports available resources, receives launch instructions, and launches/monitors local application processes such as Executors and, in cluster mode, the Driver.</p>
<p style="margin:0 0 8px;line-height:1.65">There is not one global Worker daemon for every Worker node. In a typical setup, each Worker instance runs its own Worker daemon; a single physical/virtual host can run multiple roles in a small test setup, but production layouts commonly separate the Master role from Worker hosts.</p>
<p style="margin:0;line-height:1.65"><strong>Terminology:</strong> a daemon is a kind of process, and it provides a service. Standalone has no YARN NodeManager; the Worker daemon performs the relevant local Spark process-launch/monitoring role, but is not a NodeManager.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q26. How does Spark on YARN execute in client mode versus cluster mode? Show the complete numbered architecture, lifecycle, communications, planning, scheduling, and task execution.</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">Both modes use YARN for cluster resource allocation and Spark for distributed computation. The decisive difference is the <strong>location of the Spark Driver</strong>: in client mode it runs in the submitting client process; in cluster mode it runs in the application-specific ApplicationMaster (AM) process/container. The ResourceManager (RM) is a long-running YARN service whose Scheduler and ApplicationsManager are components inside the RM service/process.</p>

<h4 style="color:#174f83;font-size:16px;margin:14px 0 8px">D. Numbered lifecycle — from submission to cleanup</h4><p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Sequencing clarification:</strong> Driver initialization must occur before the Driver can request or coordinate Executor resources. However, a query-specific physical plan does <em>not</em> have to be completed before the initial Executor request. Executor provisioning and application/query planning can overlap; Spark must have registered Executors available before it can dispatch tasks to them.</p>
<ol style="margin:0 0 10px;padding-left:22px"><li style="margin-bottom:5px"><strong>Submit:</strong> spark-submit submits the application resource reference and configuration. YARN localizes the resources needed by the AM/Driver launch; extra JARs and dependencies must be specified or made accessible as required.</li><li style="margin-bottom:5px"><strong>Accept:</strong> RM's ApplicationsManager accepts the application and coordinates AM startup.</li><li style="margin-bottom:5px"><strong>Allocate and launch AM:</strong> RM Scheduler selects a worker node; that node's NodeManager launches the AM container/process.</li><li style="margin-bottom:5px"><strong>Initialize Driver and Spark:</strong> in client mode the Driver runs in the client process; in cluster mode it runs in the AM process context. The Driver initializes Spark and its scheduling/resource-allocation infrastructure.</li><li style="margin-bottom:5px"><strong>Request Executor resources:</strong> Spark-YARN integration/AM requests Executor container resources from RM. This can happen before a particular query plan is prepared; exact timing depends on application startup and configuration.</li><li style="margin-bottom:5px"><strong>Allocate and launch Executors:</strong> RM Scheduler allocates containers to selected nodes; the AM arranges launches through the selected NodeManagers.</li><li style="margin-bottom:5px"><strong>Register Executors:</strong> Executor processes start and register with the Driver, becoming available for tasks.</li><li style="margin-bottom:5px"><strong>Plan work when triggered:</strong> when application code invokes an action, Driver-side planning and DAGScheduler create the execution plan, Jobs and Stages. This step can overlap with Executor startup and is not a prerequisite for the initial Executor request.</li><li style="margin-bottom:5px"><strong>Schedule and execute:</strong> TaskScheduler assigns tasks to available Executors; Executor processes work on partitions, report status, and exchange shuffle data when required.</li><li><strong>Finish:</strong> Driver tracks completion; Spark/YARN releases application containers and resources.</li></ol>
<h4 style="color:#174f83;font-size:16px;margin:14px 0 8px">E. Relationship and terminology checks</h4>
<ul style="margin:0 0 8px;padding-left:22px"><li><strong>NodeManager and container:</strong> the NM runs on a worker node and launches/monitors containers on that same node. The container is not a child service inside the NM; it is a resource allocation/launch context managed by the NM.</li><li><strong>Container and process:</strong> an allocated AM container hosts the AM process; an Executor container normally hosts an Executor process. Container allocation precedes process launch.</li><li><strong>RM components:</strong> Scheduler and ApplicationsManager are components within the RM service/process, not separate daemons on each worker node.</li><li><strong>RM Scheduler vs Spark TaskScheduler:</strong> RM Scheduler allocates YARN containers and selects nodes. Spark TaskScheduler assigns Spark Tasks to Executors. They operate at different layers.</li><li><strong>AM vs Driver:</strong> client mode has a separate Driver process and AM process. Cluster mode hosts the Driver within the AM process context; their responsibilities remain distinct conceptually.</li><li><strong>Physical machine, VM, and YARN node:</strong> a YARN worker node can be a physical server or a VM. A hypervisor allocates underlying physical resources to VMs; YARN manages resources visible within its registered nodes.</li><li><strong>VM vs container:</strong> a VM runs a guest OS and is managed by a hypervisor. A YARN container is primarily a per-node resource allocation and launch context, not inherently a Docker/Kubernetes container.</li><li><strong>Resource negotiation vs spilling:</strong> negotiation obtains YARN CPU/memory allocations; spilling occurs during Spark execution when intermediate data is written to disk because it cannot all remain in memory.</li></ul>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q27. In YARN, who launches the Driver and how is the application resource made available?</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>Cluster mode:</strong> the YARN ResourceManager accepts the submission and coordinates ApplicationMaster (AM) launch. The selected node's NodeManager launches the AM container/process; the Spark Driver runs in the AM process context. YARN localizes the application resources required by the launch/container, subject to the submission options and resource locations.</p>
<p style="margin:0 0 8px;line-height:1.65"><strong>Client mode:</strong> the Driver runs in the submitting client process, so it loads the application from the client-side resource. The AM still exists to request resources from YARN. Code/dependencies needed by Executors must also be made accessible through Spark/YARN's supported distribution mechanisms.</p>
<p style="margin:0;line-height:1.65">A local file path on the submitting machine is not automatically a path on every cluster node. In cluster mode, use a supported application resource/staging path and explicitly provide extra dependencies (for example, with <code>--jars</code>) when required. See <a href="https://spark.apache.org/docs/latest/running-on-yarn.html" rel="noopener noreferrer" target="_blank">Running Spark on YARN</a>.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q28. What does the YARN submission command configure, and what do policies/defaults contribute?</h3>
<p style="margin:0 0 8px;line-height:1.65">A command such as <code>spark-submit --master yarn --deploy-mode cluster --executor-memory 4g --executor-cores 2 app.jar</code> supplies the target deployment environment, Driver placement mode, resource settings, and application resource. It does not describe the complete query plan.</p>
<p style="margin:0 0 8px;line-height:1.65">The effective settings can also be influenced by Spark defaults/configuration files, Hadoop/YARN configuration, queue capacity and scheduling policies, cluster resource availability, and administrator-enforced limits. Exact precedence and enforcement depend on the relevant setting and platform.</p>
<p style="margin:0;line-height:1.65"><strong>Keep separate:</strong> requested Spark resources; YARN's actual container allocations; and Spark's per-query execution plan.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q29. Is Executor allocation before query planning an exceptional case, or can it happen in Spark Standalone, YARN, and Kubernetes?</h3>
<p style="margin:0 0 8px"><strong>It is not inherently an exceptional case.</strong> In all three deployment environments, Spark can initialize its Driver-side infrastructure and begin obtaining or launching Executors before a particular query's physical execution plan has been prepared. Resource provisioning and query planning are related but distinct activities, and their timing can overlap. The exact order depends on deployment mode, Spark version, resource-allocation configuration (including dynamic allocation), and when the application triggers an action.</p>
<p style="margin:0 0 8px">The important sequencing constraints are: the Driver and Spark runtime must be initialized sufficiently to coordinate Executor resources; the Driver must have usable registered Executors before it can dispatch tasks to them; and a query's Jobs/Stages are planned when application code triggers execution. Do not assume that the complete physical plan for a specific query must always exist before the first Executor request.</p>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q30. What are the physical machine, hypervisor, VM, YARN node, and YARN container, and how are they related?</h3>
<p style="margin:0 0 8px">A physical server supplies hardware resources. A hypervisor virtualizes that hardware into VMs. Each VM runs a guest OS and processes; when registered with YARN as a worker, it acts as a YARN node. On that node, the NodeManager launches and monitors YARN containers, which represent allocated resources and a launch context for application processes.</p>
<div style="background:#eff6ff;border:1px solid #bfdbfe;border-radius:8px;padding:10px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Physical server (hardware: CPU, RAM, storage, network)</strong><div style="margin:5px 0;color:#0875a8;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ hypervisor shares physical resources</div><div style="display:flex;flex-wrap:wrap;gap:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><div style="flex:1 1 150px;background:#fff;border:1px solid #93c5fd;border-radius:5px;padding:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>VM 1 + guest OS</strong><br/>YARN worker node<br/>NodeManager process<br/>YARN containers → application processes</div><div style="flex:1 1 150px;background:#fff;border:1px solid #93c5fd;border-radius:5px;padding:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>VM 2 + guest OS</strong><br/>May be another YARN node<br/>NodeManager process<br/>YARN containers → application processes</div></div></div>
<p style="margin:8px 0 0"><strong>Key distinction:</strong> the VM is an environment with its own guest OS; the YARN container is not another VM. Multiple VMs may share a physical server, and multiple YARN containers may run on one registered node if resource limits permit.</p>

<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q31. Why use VMs if they still consume the physical machine's CPU and memory?</h3>
<p style="margin:0 0 8px">Virtualization does not create additional physical capacity or automatically make a Spark application faster. Its value is that one physical server can host multiple isolated server environments, improving consolidation and utilization while allowing separate operating systems, configurations, resource allocations, and administrative boundaries.</p>
<p style="margin:0 0 8px"><strong>Example:</strong> a server with 32 physical cores and 128 GB RAM could host four VMs configured with 8 vCPUs and 32 GB RAM each, subject to hypervisor overhead and actual resource limits. Those VM allocations are backed by the same physical server; they do not guarantee four times the compute capacity. If all VMs demand CPU or memory simultaneously, they still compete for the underlying hardware.</p>
<p style="margin:0"><strong>Layer responsibilities:</strong> hypervisor → allocates/virtualizes physical resources for VMs; YARN RM → allocates containers on registered YARN nodes; NodeManager → launches and monitors containers on its node; Spark Driver → schedules Spark tasks; Executors → execute those tasks.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q32. What exactly does the YARN ResourceManager do, and what are the Scheduler and ApplicationsManager?</h3>
<p style="margin:0 0 8px">The ResourceManager is a cluster-level YARN service/process. Its <strong>Scheduler</strong> allocates resources according to available capacity, queues, policies, and requests; it selects nodes for container allocations. The <strong>ApplicationsManager</strong> accepts application submissions and coordinates the initial ApplicationMaster launch and application lifecycle handling. They are components of the RM, not separate services running on every worker node.</p>
<p style="margin:0 0 8px">The RM receives node status and resource information through NodeManager heartbeats. It allocates containers, but the selected NodeManager launches and monitors each container locally. The ApplicationMaster communicates with RM to request and receive allocations and coordinates launching those allocations through the selected NodeManagers.</p>
<p style="margin:0"><strong>Do not confuse the schedulers:</strong> YARN's RM Scheduler chooses nodes and allocates containers; Spark's Driver-side DAGScheduler and TaskScheduler organize Jobs/Stages and assign Tasks to Executors.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q33. What is the difference between a YARN container, a Docker container, and a Kubernetes container?</h3>
<p style="margin:0 0 8px">A <strong>YARN container</strong> is primarily a resource allocation (such as memory and vcores) plus a launch context on a specific YARN node; the NodeManager launches and monitors the process associated with it. A <strong>Docker container</strong> is an OS-level isolated runtime created from an image. A <strong>Kubernetes container</strong> is a container runtime instance managed as part of a Pod by Kubernetes; a Pod can contain one or more containers.</p>
<p style="margin:0 0 8px">These uses of the word “container” are not equivalent. A YARN container does not inherently imply Docker or an image-based filesystem. YARN can be configured to launch workloads with a Docker runtime, but that is an optional integration rather than the definition of a YARN container.</p>
<p style="margin:0"><strong>Conceptual contrast:</strong> YARN container = resource allocation/launch context; Docker container = OS-level runtime instance; Kubernetes Pod = scheduling unit that groups container(s) and related configuration.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q34. What is the difference between resource negotiation, container allocation, container launch, Executor registration, and task execution?</h3>
<ol style="margin:0;padding-left:22px"><li style="margin-bottom:5px"><strong>Resource request / negotiation:</strong> the application-specific AM requests resources through RM.</li><li style="margin-bottom:5px"><strong>Allocation:</strong> RM Scheduler selects a node and allocates a container with requested resources.</li><li style="margin-bottom:5px"><strong>Launch:</strong> AM arranges for the selected node's NM to launch the allocated container; NM starts the process.</li><li style="margin-bottom:5px"><strong>Registration:</strong> an Executor process initializes and registers with the Spark Driver.</li><li><strong>Task execution:</strong> the Driver's Spark schedulers assign tasks; the Executor process performs computation on data partitions.</li></ol>
<p style="margin:8px 0 0">These are separate events. Allocation alone does not mean an Executor has started, and an Executor starting does not mean it has already received a task.</p>

<div style="text-align:center;font-size:13px;color:#526577;margin-top:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">The Driver communicates with Executors to schedule tasks and coordinate execution.</div>
</div>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Execution flow:</strong> submit application → Kubernetes creates/schedules the Driver Pod → the Driver starts and initializes SparkSession/SparkContext → the Driver requests Executor resources through Spark's Kubernetes integration → Kubernetes schedules Executor Pods → the Driver schedules Spark tasks → Executors process partitions and return status/results → Pods and related resources are cleaned up when the application ends.</p>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0"><strong>Important:</strong> The Control Plane does not execute Spark tasks. It manages the Kubernetes cluster and Pod placement. The Spark Driver coordinates Spark execution; Executor processes perform the distributed work. The diagram shows the common cluster-mode architecture.</p>

<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q36. Is the hierarchy Pod → Container → Driver/Executor correct? Is a Pod separate from the Driver and Executor?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0"><strong>Yes, with one terminology clarification:</strong> a Pod is a Kubernetes object/deployment unit; a container is the runtime environment inside the Pod; and the Spark Driver or Executor is a process that runs inside a container. So the conceptual containment relationship is:</p>
<div style="width:100%;box-sizing:border-box;background:#eff6ff;border:1px solid #bfdbfe;border-radius:8px;padding:12px;margin:10px 0;text-align:center;overflow-wrap:anywhere;min-width:0;max-width:100%">
<div style="box-sizing:border-box;background:#ffffff;border:1px solid #b9d9f5;border-radius:7px;padding:11px;color:#174f83;overflow-wrap:anywhere;min-width:0;max-width:100%"><strong>Pod</strong><div style="font-size:13px;color:#526577;margin-top:3px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Kubernetes deployable unit</div>
<div style="text-align:center;font-size:20px;color:#0875a8;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">contains ↓</div>
<div style="box-sizing:border-box;background:#e4f6e9;border:1px solid #a7f3d0;border-radius:6px;padding:10px;color:#166534;overflow-wrap:anywhere;min-width:0;max-width:100%"><strong>Container</strong><div style="font-size:13px;color:#334155;margin-top:3px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">Runs the application process</div>
<div style="text-align:center;font-size:20px;color:#0875a8;margin:3px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">runs ↓</div>
<div style="box-sizing:border-box;background:#ffffff;border:1px solid #cbd5e1;border-radius:5px;padding:9px;color:#24324b;overflow-wrap:anywhere;min-width:0;max-width:100%"><strong>Spark Driver process</strong> <em>or</em> <strong>Spark Executor process</strong></div>
</div>
</div>
</div>
<ul style="box-sizing:border-box;margin:0 0 8px;min-width:0">
<li><strong>Typical Spark-on-Kubernetes mapping:</strong> one Driver Pod contains a Driver container running the Driver process; each Executor Pod contains an Executor container running one Executor process.</li>
<li><strong>Pod to containers:</strong> a Pod contains one or more containers. A supporting sidecar may be present alongside the main Spark container.</li>
<li><strong>Container to process:</strong> a container can run a main process and may spawn child processes; for the Spark architecture model, think of the Driver or Executor as the main Spark process inside its container.</li>
</ul>
<p style="box-sizing:border-box;margin:0;overflow-wrap:anywhere;min-width:0">Therefore, the Pod is separate from the Driver/Executor as a Kubernetes abstraction, but it hosts the container in which the Driver/Executor process runs. The Driver and Executor are Spark roles/processes, not Kubernetes object types.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q37. Explain Kubernetes Architecture and Spark Execution in both Client and Cluster deployment modes. Identify every major component, its type, containment relationships, configuration owner, and communication flow.</h3>
<p style="margin:0 0 10px;line-height:1.6">Kubernetes is a container-orchestration platform. It places Pods on worker nodes and maintains their desired state. Spark coordinates its own application: the Driver plans jobs/stages and schedules Spark tasks; Executor processes execute those tasks. The Kubernetes Scheduler schedules <em>Pods</em>, not Spark tasks. The diagrams below show the two deployment modes separately.</p>
<h4 style="color:#174f83;margin:12px 0 8px">A. Kubernetes client mode — Driver runs in the submitting client process</h4>
<div style="width:100%;box-sizing:border-box;background:#eff6ff;border:1px solid #bfdbfe;border-radius:8px;padding:12px;margin:8px 0;overflow-wrap:anywhere;min-width:0;max-width:100%">
<div style="background:#fff;border:1px solid #93c5fd;border-radius:7px;padding:9px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#174f83">1. Developer / spark-submit client</strong><br/><span style="font-size:13px;color:#526577">Process on a client computer or edge node; submits application and options. In client mode, the Spark Driver runs in this client-side process environment.</span></div>
<div style="text-align:center;color:#0875a8;font-size:17px;margin:5px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ authenticated Kubernetes API requests / Pod specifications</div>
<div style="background:#fff;border:1px solid #93c5fd;border-radius:7px;padding:9px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#174f83">2. Kubernetes API Server</strong> <span style="font-size:12px;color:#526577">[Control Plane service/process]</span><br/><span style="font-size:13px;color:#526577">Validates API requests and is the primary interface used by clients and cluster components.</span></div>
<div style="text-align:center;color:#0875a8;font-size:17px;margin:5px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↕ API Server reads/writes cluster state in etcd; components watch API state</div>
<div style="background:#fff;border:1px solid #93c5fd;border-radius:7px;padding:9px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#174f83">3. Kubernetes Control Plane</strong> <span style="font-size:12px;color:#526577">[Logical management layer; services/processes]</span><div style="display:flex;flex-wrap:wrap;gap:7px;margin-top:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>etcd</strong><br/>Distributed key-value database storing cluster state; API Server interacts with it.</div><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>kube-scheduler</strong><br/>Chooses a worker node for each unscheduled Pod based on resources and constraints.</div><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Controller manager</strong><br/>Controllers reconcile actual state toward desired state through the API Server.</div></div></div>
<div style="text-align:center;color:#0875a8;font-size:17px;margin:5px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ scheduler records Pod-to-node binding via API Server; node agent observes assigned Pod</div>
<div style="background:#fff;border:1px solid #93c5fd;border-radius:7px;padding:9px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#174f83">4. Worker node(s)</strong> <span style="font-size:12px;color:#526577">[Physical machine or VM running a node OS]</span><br/><span style="font-size:13px;color:#526577">Each node has node-level software and contributes CPU, RAM, storage and networking.</span><div style="display:flex;flex-wrap:wrap;gap:7px;margin-top:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><div style="flex:1 1 180px;background:#fef3c7;border:1px solid #fcd34d;border-radius:5px;padding:8px;font-size:12px;color:#78350f;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>kubelet</strong> [node-agent process]<br/>Talks to API Server, observes Pods assigned to this node, and asks the runtime to make their containers run.</div><div style="flex:1 1 180px;background:#fef3c7;border:1px solid #fcd34d;border-radius:5px;padding:8px;font-size:12px;color:#78350f;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Container runtime</strong> [node software/service, e.g. containerd or CRI-O]<br/>Pulls images and creates, starts, monitors and stops containers. <div style="text-align:center;color:#0875a8;font-size:17px;margin:5px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↔ Driver process on client ↔ Executor processes in Pods; Driver schedules Spark tasks</div>
<div style="background:#fff;border:1px solid #93c5fd;border-radius:7px;padding:9px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#174f83">7. Spark execution layer</strong><br/><span style="font-size:13px;color:#526577">Client-side Driver plans jobs/stages and schedules tasks; Executor processes in Kubernetes execute tasks. </div>
<h4 style="color:#174f83;margin:14px 0 8px">B. Kubernetes cluster mode — Driver runs inside a Driver Pod</h4>
<div style="width:100%;box-sizing:border-box;background:#ecfdf5;border:1px solid #a7f3d0;border-radius:8px;padding:12px;margin:8px 0;overflow-wrap:anywhere;min-width:0;max-width:100%">
<div style="background:#fff;border:1px solid #86efac;border-radius:7px;padding:9px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#166534">1. Developer / spark-submit client</strong><br/><span style="font-size:13px;color:#526577">Submits the application resource reference and configuration; it does not host the long-running Driver in cluster mode. <div style="text-align:center;color:#047857;font-size:17px;margin:5px 0;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere">↓ submits Driver Pod specification through Kubernetes API</div>
<div style="background:#fff;border:1px solid #86efac;border-radius:7px;padding:9px;text-align:center;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong style="color:#166534">2. API Server ↔ etcd</strong> <span style="font-size:12px;color:#526577">[Control Plane API service + state database]</span><br/><span style="font-size:13px;color:#526577">API Server validates requests and persists cluster state in etcd.</span></div>

Kubernetes Control Plane</strong> <span style="font-size:12px;color:#526577">[logical layer, not one single process]</span><div style="display:flex;flex-wrap:wrap;gap:7px;margin-top:8px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>kube-scheduler</strong><br/>Binds each pending Pod to a suitable node; it does not schedule Spark tasks.</div><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Controller manager</strong><br/>Reconciles managed resources; it is not the Spark application scheduler.</div><div style="flex:1 1 150px;background:#f8fafc;border:1px solid #cbd5e1;border-radius:5px;padding:8px;font-size:12px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>API Server</strong><br/>All these components coordinate through Kubernetes API state; etcd is accessed by API Server.</div></div></div>

<h4 style="color:#174f83;margin:14px 0 8px">End-to-end communication and execution sequence</h4>
<ol style="margin:0 0 10px;padding-left:22px;line-height:1.55">
<li>Developer configures the application resource, container image, Kubernetes credentials/namespace, deploy mode and Spark resource settings, then invokes <code>spark-submit</code>. The Driver Pod must be able to access the application code: a <code>local://</code> JAR is expected to be in the image, while other supported resource URIs must be reachable.</li>
<li><strong>Client mode:</strong> Driver remains in the client-side process. <strong>Cluster mode:</strong> Spark submits a Driver Pod specification through the Kubernetes API Server, and the Driver starts inside its container.</li>
<li>API Server validates requests and stores cluster state in etcd. Kubernetes Control Plane components communicate through the API Server; controllers reconcile state and kube-scheduler selects nodes for unscheduled Pods.</li>
<li>After a Pod is assigned to a node, that node's kubelet observes the assignment through the API Server and asks the container runtime to pull the image and start the container via CRI.</li>
<li>The Driver initializes Spark, plans work into jobs/stages/tasks, and uses Spark's Kubernetes integration to request Executor Pods through the Kubernetes API.</li>
<li>Kubernetes schedules each Executor Pod to a node. The selected node's kubelet and runtime start its container; the Executor process initializes and registers with the Driver.</li>
<li>The Spark Driver's task scheduler sends tasks to registered Executors. Executors process partitions and communicate task status/results back to the Driver over Spark's application network connections.</li>
<li>When the application ends, Spark and Kubernetes clean up application resources according to Spark settings and cluster policies. Kubernetes Pod scheduling and lifecycle management remain separate from Spark's task scheduling.</li>
</ol>
<h4 style="color:#174f83;margin:14px 0 8px">Important distinctions</h4>
<ul style="margin:0 0 10px;padding-left:22px;line-height:1.55">
<li><strong>Pod contains container(s); container runs process(es).</strong> A Pod is the scheduling unit, not a process. A container is a runtime environment, not a VM.</li>
<li><strong>kubelet is essential:</strong> it runs on each worker node, watches Pods assigned to that node, and coordinates with the container runtime. It does not choose the node; kube-scheduler does.</li>
<li><strong>Kubernetes Scheduler vs Spark scheduler:</strong> Kubernetes places Pods on nodes; the Spark Driver schedules Spark tasks onto Executors.</li>
<li><strong>API Server vs etcd:</strong> components and clients use the API Server; etcd stores cluster state and is normally accessed through the API Server.</li>
<li><strong>Client vs cluster mode:</strong> the primary difference is Driver location. In client mode the Driver can run on a client physical host, VM, edge/gateway host, or supported Pod setup; in cluster mode it runs in a Driver Pod. In both modes, Executor Pods run in Kubernetes and the Driver coordinates Spark task execution.</li>
<li><strong>Environment vs process:</strong> the node OS and container filesystem/runtime are environments; kubelet, Driver and Executor are processes; Pod is a Kubernetes object; image is a package; physical machine is hardware; VM is a virtual computer.</li>
<li><strong>Network communication:</strong> API requests/control-state traffic use the Kubernetes API; kubelet-to-runtime calls use CRI; Driver-to-Executor traffic uses Spark's application protocols over Pod/client networking. Exact ports and connectivity depend on Spark and cluster configuration.</li>
</ul>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q38. In Kubernetes cluster mode, how does the Driver Pod obtain the submitted application code?</h3>
<p style="margin:0 0 8px;line-height:1.65">The submission process asks the Kubernetes API Server to create a Driver Pod specification. Kubernetes schedules the Pod to a node; that node's kubelet asks the container runtime to start the container. The Driver process then loads the application resource from the location configured for the submission.</p>
<p style="margin:0 0 8px;line-height:1.65">The resource must actually be accessible inside the Driver environment. For example, a <code>local://</code> JAR URI is intended for a JAR already present in the container image; other supported URI schemes can reference remote resources that the Driver can fetch. Do not assume a local path on the submitting laptop is automatically present inside the Driver Pod.</p>
<p style="margin:0;line-height:1.65">See <a href="https://spark.apache.org/docs/latest/running-on-kubernetes.html" rel="noopener noreferrer" target="_blank">Running Spark on Kubernetes</a>.</p>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q39. In Kubernetes client mode, must the Driver run on a Kubernetes Worker node?</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>No.</strong> In client mode, the Driver can run in the submitting process on a physical host, VM, gateway/edge machine, or—in supported setups—inside a Pod. It does not have to be on a Kubernetes worker node. Kubernetes still manages the Executor Pods.</p>
<p style="margin:0 0 8px;line-height:1.65">The essential requirement is network reachability: Executor Pods must be able to connect to the Driver using a routable hostname/address and port. If the Driver is outside Kubernetes, routing, firewall rules, authentication, and advertised Driver host/port settings must be configured appropriately.</p>
<p style="margin:0;line-height:1.65">The Kubernetes Scheduler places Pods; it does not place or schedule a Driver process that remains outside Kubernetes in client mode.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q40. In Kubernetes, who plans Spark queries: the Control Plane or the Spark Driver?</h3>
<p style="margin:0 0 8px;line-height:1.65">The <strong>Spark Driver</strong> plans the Spark computation and schedules Spark Tasks. The Kubernetes Control Plane manages Kubernetes objects and Pod placement: the API Server exposes the API, etcd stores cluster state via the API Server, controllers reconcile state, and kube-scheduler selects nodes for Pods.</p>
<p style="margin:0;line-height:1.65">Kubernetes scheduling a Driver or Executor Pod is not the same as Spark scheduling a Task. After an Executor Pod starts, the Executor process registers with the Driver; the Driver's TaskScheduler then dispatches Spark Tasks to registered Executors.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q42. What is the difference between a Job, Stage, and Task, and who schedules or executes each one?</h3>
<p style="box-sizing:border-box;margin:0 0 8px;overflow-wrap:anywhere;min-width:0">A <strong>Job</strong> is a unit of computation commonly triggered by an action such as <code>count()</code> or <code>collect()</code>. The Driver's <strong>DAGScheduler</strong> handles the Job and divides its execution into <strong>Stages</strong> according to dependencies, especially shuffle boundaries. The Driver's <strong>TaskScheduler</strong> schedules individual <strong>Tasks</strong> for a stage onto available Executor task slots. An Executor process executes each task, typically processing one partition for that stage. A Job is therefore not executed by one machine or process; its stages and tasks collectively perform the work across Executors.</p>
<ul style="margin:0;padding-left:22px;overflow-wrap:anywhere">
<li><strong>Driver:</strong> coordinates the application and schedules work.</li>
<li><strong>DAGScheduler:</strong> manages Jobs and divides them into Stages.</li>
<li><strong>TaskScheduler:</strong> coordinates scheduling Tasks on Executors.</li>
<li><strong>Executor:</strong> executes Tasks and processes data partitions.</li>
</ul>
<p style="box-sizing:border-box;margin:8px 0 0;overflow-wrap:anywhere;min-width:0">Example: <code>df.filter("id % 2 = 0").count()</code> commonly triggers a Job. If the relevant stage has four partitions, Spark may create four Tasks for that stage. The exact Job, Stage and Task counts depend on the execution plan, partitioning, shuffle operations and runtime settings.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q43. How is the Driver computation plan different from resource configuration?</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>Resource configuration</strong> specifies or requests capacity and settings—for example, Executor memory, Executor cores, and initial Executor count. The effective outcome may be constrained by platform availability and policies/defaults.</p>
<p style="margin:0 0 8px;line-height:1.65"><strong>Computation/execution plan</strong> describes the operations and dependencies Spark will execute: operators, Jobs, Stages, shuffle boundaries, and Tasks. It is built from application operations and Spark's planning/optimization logic, not simply read from <code>spark-submit</code> resource settings.</p>
<p style="margin:0 0 8px;line-height:1.65"><strong>Runtime behavior</strong> is what actually happens during execution, including real data volumes, task durations, memory pressure, and spilling. Resource provisioning and planning can overlap; exact timing depends on deploy mode and configuration.</p>
<div style="background:#eff6ff;border:1px solid #bfdbfe;border-radius:6px;padding:10px;box-sizing:border-box;max-width:100%;min-width:0;overflow-wrap:anywhere"><strong>Mental model:</strong> configuration = requested capacity; cluster manager = allocation/placement; Driver plan = computation to perform; TaskScheduler = task dispatch; Executors = distributed task execution.</div>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q44. Who determines how much data each Task processes and how many Tasks run at once?</h3>
<p style="margin:0 0 8px;line-height:1.65">There is no single setting that directly determines both values. Input file layout and read-splitting rules influence input partitions; shuffle settings and the execution plan influence later-stage partitions; one task normally processes one partition for a stage task. Actual partition sizes can vary.</p>
<p style="margin:0 0 8px;line-height:1.65">Concurrent Tasks are constrained by available Executor task slots (influenced by Executor cores and Spark scheduling) and by resources actually allocated. If a stage has 12 partitions and 6 available task slots, it may run in two waves of six Tasks, assuming one Task per partition and no other scheduling constraints.</p>
<p style="margin:0;line-height:1.65">The Driver determines the work units and schedules Tasks; the cluster manager allocates process resources; partitioning determines the work size; runtime conditions affect how efficiently a Task can process its partition.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q45. Where can spilling occur, and can the Driver plan predict the exact amount?</h3>
<p style="margin:0 0 8px;line-height:1.65">Spilling is most commonly discussed for Executor-side task operators—such as sorting, aggregation, and shuffle processing—that write intermediate data to disk when it cannot all be kept in execution memory. It is a runtime response to actual data sizes and memory pressure, not simply a fixed value in the Driver's query plan.</p>
<p style="margin:0 0 8px;line-height:1.65">The plan may indicate operations that can require memory-intensive processing, but it does not generally predict the exact amount of spill that will occur. The Driver can also perform its own local work, but that is distinct from spilling generated by distributed Executor tasks.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q46. Can a client-mode Driver host also be a Worker node? Are Master and Worker nodes physical machines or VMs?</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>Yes, where the deployment supports it.</strong> In Standalone client mode, the Driver runs on the submitting host, and that host may also run a Worker service; however, it does not have to be a configured Worker. In YARN client mode, the Driver runs in the client process and that host need not be a YARN worker node. In Kubernetes client mode, the Driver can run outside Kubernetes or in a Pod.</p>
<p style="margin:0 0 8px;line-height:1.65">A Master/Worker/node is a logical platform role, not a hardware type. It can run on a physical server or a VM. A common enterprise practice is to run cluster-management services on dedicated or separately managed hosts and distributed workloads on a pool of worker VMs or cloud instances, though small labs may combine roles.</p>
<p style="margin:0;line-height:1.65">A Standalone Master host runs the Spark Master service; Standalone Worker hosts run Worker daemons. YARN has ResourceManager services and NodeManagers on worker nodes. Kubernetes has Control Plane components and worker nodes running kubelet/container runtime. These are platform-specific roles, not interchangeable service names.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q47. What is the difference between application code, query planning, and distributed data processing?</h3>
<p style="margin:0 0 8px;line-height:1.65">The submitted application code is the program the Driver starts. The Driver executes the application entry point and orchestrates Spark APIs. When an action triggers computation, Spark plans Jobs/Stages/Tasks. Executors then execute those Tasks over data partitions.</p>
<p style="margin:0 0 8px;line-height:1.65">For example, <code>read → groupBy(region) → count()</code> is not copied as three whole stages to three nodes. Instead, Spark builds a plan; Tasks process partitions in parallel; a shuffle may redistribute records so matching keys can be aggregated; subsequent Tasks process the resulting partitions. A 3 GB input is not automatically divided into three 1 GB partitions.</p>
<p style="margin:0;line-height:1.65"><strong>Remember:</strong> code starts in the Driver; work is represented by Spark Tasks; distributed data processing happens on Executor processes, and the partition layout can change across shuffle stages.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q48. What is the difference between a process, a service, and a daemon?</h3>
<p style="margin:0 0 8px;line-height:1.65"><strong>Process:</strong> a running instance of a program. <strong>Service:</strong> describes a function provided to other components. <strong>Daemon:</strong> emphasizes a long-running background process that provides a service without direct interactive use.</p>
<p style="margin:0;line-height:1.65">A Standalone Worker is a process that acts as a service and is commonly called a Worker daemon. The Standalone Master is also a long-running process/service. In YARN, the NodeManager is a node-level daemon; in Kubernetes, kubelet is a node agent. Similar responsibilities do not make these components identical.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q49. Is Spark RPC specific to Kubernetes, or is it used with Standalone and YARN as well?</h3>
<p style="margin:0;line-height:1.7">Spark's internal RPC mechanism is part of Spark's distributed runtime, not a Kubernetes-only feature. Spark uses RPC for communication between processes such as the Driver and Executors in Standalone, YARN, and Kubernetes deployments. In the standard networking implementation, Spark uses Netty over TCP (Transmission Control Protocol), a reliable, ordered transport protocol. The cluster manager determines how Spark processes are deployed and resources are allocated; it does not replace Spark's Driver–Executor communication mechanism. Kubernetes API communication is separate from Spark RPC.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q50. What are kubectl and kubelet, and where do they run?</h3>
<p style="margin:0;line-height:1.7"><strong>kubectl</strong> is the Kubernetes command-line client. It commonly runs on an administrator's workstation or another client machine and sends requests to the Kubernetes API Server. <strong>kubelet</strong> is an agent that runs on each Kubernetes Worker Node and ensures that Pods assigned to that Node are running. The normal control path is kubectl → API Server → Kubernetes control-plane components; kubelet communicates with the API Server and the local Container Runtime. kubectl and kubelet are different components.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q51. What are Node taints, Pod tolerations, and Node affinity?</h3>
<p style="margin:0;line-height:1.7">A <strong>taint</strong> is a restriction placed on a Node to repel Pods that do not tolerate it. A <strong>toleration</strong> is declared on a Pod and allows it to be considered for a Node with a matching taint; it does not attract the Pod or guarantee placement. <strong>Node affinity</strong> is a Pod scheduling rule that selects Nodes based on their labels. For example, a Node may have label <code>hardware=gpu</code> and taint <code>dedicated=gpu:NoSchedule</code>. A GPU Pod can require the label through node affinity and tolerate the taint through a toleration. Taints repel; tolerations permit; affinity selects or prefers.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q52. What do the Kubernetes taint effects NoSchedule, PreferNoSchedule, and NoExecute mean?</h3>
<p style="margin:0;line-height:1.7"><strong>NoSchedule</strong> prevents new Pods without a matching toleration from being scheduled on the Node; already-running Pods are not evicted just because this taint is added. <strong>PreferNoSchedule</strong> tells the Scheduler to avoid the Node when possible, but it is not an absolute prohibition. <strong>NoExecute</strong> prevents new non-tolerating Pods from being scheduled and can evict existing non-tolerating Pods. A Pod can use <code>tolerationSeconds</code> with a NoExecute toleration to remain for a specified period before eviction. Therefore, changing a Node's taint after a Pod has been scheduled can affect that already-running Pod when the effect is NoExecute.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q53. How does the Kubernetes Scheduler select a Worker Node, and who configures its policy?</h3>
<p style="margin:0;line-height:1.7">The Kubernetes Scheduler performs <strong>filtering</strong> to identify Nodes that satisfy the Pod's resource requests and scheduling constraints, then <strong>scoring</strong> to rank eligible Nodes. For a Pod requesting 4 CPU cores, a Node with only 2 allocatable unrequested cores is not eligible, while Nodes with 4 and 8 may both be eligible. The Scheduler does not automatically choose the exact-fit Node; scoring plugins and scheduling constraints determine the ranking. Kubernetes supplies default scheduling behaviour. Cluster administrators can configure Scheduler profiles and plugins, while application teams specify Pod resource requests, affinity, tolerations, and other workload requirements. The Scheduler makes the placement decision.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q54. Are the Deployment, ReplicaSet, Node, Job, and EndpointSlice controllers components of Controller Manager?</h3>
<p style="margin:0;line-height:1.7">Yes. These are examples of individual Kubernetes controllers that run under the <strong>kube-controller-manager</strong> component in the Control Plane. Each controller runs a reconciliation loop: it compares desired state with observed state and takes action to reduce the difference. For example, the Deployment Controller manages the Deployment's desired state and its ReplicaSets; the ReplicaSet Controller maintains the requested number of Pods. The Scheduler selects a Node for an unscheduled Pod, and kubelet starts its Containers. Controller Manager does not itself schedule Pods or start Containers.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q55. Does etcd record every Spark task completion or Executor task status?</h3>
<p style="margin:0;line-height:1.7">No. etcd is the persistent key-value datastore for Kubernetes cluster state; it is not Spark's task-status database. When a Spark Executor completes a task, it reports task status to the Spark Driver through Spark's internal communication mechanisms, including Spark RPC. Kubernetes does not need an etcd write for every Spark task completion. Kubernetes resource state is updated when Kubernetes-relevant events occur, such as Pod status changes, Pod termination, or resource deletion. An Executor may finish its currently assigned tasks and remain alive in a Running Pod to execute later tasks.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q56. What is etcd, and how long does Kubernetes keep its information?</h3>
<p style="margin:0;line-height:1.7">etcd is a distributed, consistent key-value datastore used by Kubernetes to persist cluster state, such as Kubernetes resource objects and metadata. It is not a relational SQL database, a document database like MongoDB, or a Spark data/shuffle store. Its data normally persists across individual Pod and application restarts and for the lifetime of the cluster, unless resources are deleted or the datastore is reset or lost. Kubernetes components normally access this state through the API Server rather than writing directly to etcd. Production cluster operations must include etcd backup and recovery planning.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q57. Does one Spark Executor normally run in one Kubernetes Pod? Can a Pod contain multiple Executors?</h3>
<p style="margin:0;line-height:1.7">In Spark's standard Kubernetes deployment model, each Spark Executor instance is launched in its own Executor Pod, normally with one primary Executor Container running one Executor process. Thus, <code>spark.executor.instances=3</code> normally results in three Executor Pods. Kubernetes itself permits multiple Containers in a Pod, but that does not mean Spark automatically places multiple Executor processes in one Pod. Additional sidecar Containers can be added using Spark Kubernetes Pod templates; those sidecars do not become additional Spark Executors. This one-Executor-per-Pod mapping is implemented by Spark's Kubernetes backend, not manually specified as a separate mapping by the user.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q58. How are Spark resource configurations different from Kubernetes Pod resource configurations?</h3>
<p style="margin:0;line-height:1.7">They control related but different layers. Spark settings such as <code>spark.executor.instances</code>, <code>spark.executor.cores</code>, and <code>spark.executor.memory</code> describe the number of Spark Executors and their execution/JVM resource configuration. Kubernetes settings and Pod specifications describe Pod-level resource requests and limits used for scheduling and runtime enforcement, along with deployment details such as namespace, image, service account, labels, affinity, and tolerations. Executor memory overhead also contributes to the Pod's memory requirement. Spark configuration and Kubernetes resource configuration must be planned together; a Spark core count is not automatically identical to a Kubernetes CPU request or limit.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q59. Can Kubernetes run applications other than Spark, and why is it studied as a Spark cluster manager?</h3>
<p style="margin:0;line-height:1.7">Kubernetes is a general-purpose container orchestration platform. It can deploy and manage many containerized workloads, including APIs, web applications, databases, message-processing services, ML inference services, and Spark applications. It is studied in Spark deployment because Spark supports Kubernetes as one of its cluster-manager/deployment backends. In that context, Kubernetes schedules Driver and Executor Pods and manages their Container lifecycle. Studying Spark on Kubernetes explains one way Spark uses the platform; broader Kubernetes study covers its general architecture, networking, storage, security, and operations.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q60. How do Spark Standalone, YARN, and Kubernetes differ as deployment and resource-management options?</h3>
<p style="margin:0;line-height:1.7"><strong>Spark Standalone</strong> uses Spark's own Master and Worker services and is suited to relatively straightforward Spark-focused clusters. <strong>YARN</strong> uses Hadoop's ResourceManager and NodeManagers to manage cluster resources and launch YARN Containers, making it relevant to Hadoop-centric environments. <strong>Kubernetes</strong> is a general container orchestration platform; its Control Plane schedules Pods and its Node agents manage Containers, allowing Spark to run alongside many other containerized workloads. The choice depends on existing infrastructure, workload mix, operational skills, security, storage/network integration, and platform strategy. Kubernetes is not automatically faster or cheaper, and YARN or Standalone are not inherently unsuitable for production.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q61. What is container orchestration, and why might a Kubernetes tutorial use Nginx?</h3>
<p style="margin:0;line-height:1.7">Container orchestration means coordinating the deployment and operation of Containers: deciding where workloads run, maintaining the requested number of instances, restarting failed workloads, exposing services, and managing resource requirements. Kubernetes automates much of this through resources such as Deployments, Pods, Services, and scheduling policies. Nginx is not only a UI development tool; it is a web server, reverse proxy, and load-balancing component. It is often used as a lightweight sample workload because it lets learners practise creating Pods, Deployments, Services, scaling, logs, and networking without first building a complex application.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q62. How can user traffic be load-balanced across multiple FastAPI instances running on Kubernetes?</h3>
<p style="margin:0;line-height:1.7">A production design can run multiple stateless FastAPI replicas managed by a Kubernetes Deployment. A Kubernetes Service provides a stable endpoint and distributes connections/traffic to matching healthy Pod endpoints. External HTTP/HTTPS traffic can enter through an Ingress controller or a cloud load balancer, depending on the architecture. A Horizontal Pod Autoscaler can adjust replica count using configured metrics. The design should also include readiness and liveness probes, resource requests and limits, timeouts, concurrency controls, rate limiting, observability, and graceful shutdown. Kubernetes Service load distribution is distinct from a full-featured external load balancer; exact behaviour depends on the networking implementation.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q63. Why does a GenAI application need a separate LLM capacity and routing strategy in addition to API load balancing?</h3>
<p style="margin:0;line-height:1.7">Balancing incoming user requests across FastAPI replicas does not automatically balance or increase capacity at the LLM provider. All API replicas may still call the same provider account or endpoint and encounter request/token rate limits, concurrency limits, latency, or provider outages. A GenAI architecture may therefore need a separate model-routing layer with provider/model selection, rate limiting, concurrency controls, bounded retries, timeouts, circuit breakers, and fallback policies. Conversation/session state should be stored in a shared store or database rather than only in one API process's local memory, so requests can be served by different replicas.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q64. What is SRE, and what responsibilities does an SRE team have?</h3>
<p style="margin:0;line-height:1.7">SRE means <strong>Site Reliability Engineering</strong>. SRE applies software engineering and operational practices to keep production services reliable, available, performant, and maintainable. Typical responsibilities include service monitoring and observability, incident response, capacity planning, performance and reliability improvements, automation, disaster recovery, and managing operational risk. SRE teams commonly work with <strong>SLIs</strong> (Service Level Indicators, measured service behaviour) and <strong>SLOs</strong> (Service Level Objectives, target levels for those indicators). For a GenAI service, useful measures include availability, latency, error rates, throughput, token consumption, and cost.</p>
</div>
<div style="width:calc(100% - 24px);box-sizing:border-box;max-width:calc(100% - 24px);min-width:0;overflow:hidden;overflow-wrap:anywhere;border:1px solid #cbd5e1;border-radius:8px;padding:14px;margin:12px auto;background:#ffffff;color:#334155">
<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q65. Can Kubernetes be practised locally, and what is the difference between local and managed cloud Kubernetes?</h3>
<p style="margin:0;line-height:1.7">Kubernetes can be practised locally; a cloud provider is not required. Tools such as Minikube and kind create local clusters, while Docker Desktop also offers local Kubernetes functionality. Local clusters support learning Pods, Deployments, Services, scheduling, taints, affinity, resource requests, and many Kubernetes API behaviours. Managed services such as Google Kubernetes Engine (GKE), Amazon EKS, and Azure Kubernetes Service (AKS) add real cloud networking, IAM integration, managed load balancers, cloud storage, node pools, autoscaling, and provider-specific operations. Managed Control Plane responsibilities are handled by the provider to varying degrees; self-managed Kubernetes on VMs requires more direct administration. Cloud resources can incur charges even when a provider offers free credits or a cluster-management fee allowance.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q66. Are Kubelet and Container Runtime present on Worker Nodes in Standalone Spark and YARN too?</h3>
<p style="margin:0;line-height:1.7">No. <strong>Kubelet and the Kubernetes Container Runtime integration are Kubernetes Node components.</strong> Similar-looking Worker Nodes in the three deployment models do not have identical software. In Spark Standalone, a Spark Worker service manages Spark Executor processes. In YARN, a NodeManager manages YARN Containers in which Spark processes can run. In Kubernetes, kubelet works with the Container Runtime through the Container Runtime Interface (CRI) to create and manage Pod Containers. A YARN Container is not a Kubernetes Pod, and a Spark Worker is not a kubelet.</p>
</div>

<h3 style="box-sizing:border-box;font-size:17px;color:#174f83;margin:0 0 7px;overflow-wrap:anywhere;min-width:0">Q67. What local Kubernetes environment can be used for hands-on practice, and what does it teach compared with cloud Kubernetes?</h3>
<p style="margin:0;line-height:1.7">A local Kubernetes cluster can be created without AWS, Google Cloud, or Azure. Common learning options include <strong>Minikube</strong>, <strong>kind</strong>, and Docker Desktop's Kubernetes feature. A practical Windows learning path is WSL2 with an Ubuntu distribution, Docker Desktop using its WSL2 backend, and Minikube using the Docker driver. A multi-node Minikube cluster can help practise scheduling and Pod placement, although its Nodes are typically containers on one physical machine rather than independent servers. Local practice covers much of the Kubernetes API, Pods, Deployments, Services, scheduling, and resource controls. Cloud platforms add provider IAM, VPC networking, managed load balancers, cloud storage, node pools, multi-zone infrastructure, managed Control Plane operations, and billing. Self-managed Kubernetes on cloud VMs is a separate path for deeper administrator practice.</p>
</div>

<p>Use this conceptual model to reason about all three cluster-manager options. It is not a strict sequence of four independent components; deployment, resource negotiation, process launch, and initialization can overlap.</p>
<p>For Kubernetes deployments, distinguish Spark Driver–Executor RPC from Kubernetes API operations. Spark's Kubernetes backend creates Driver and Executor Pods; the Kubernetes Scheduler assigns Pods to Nodes; kubelet and the Container Runtime start their Containers. Kubernetes Services and external ingress/load-balancing components handle application network traffic, while Spark RPC coordinates Spark runtime communication. Kubernetes etcd stores Kubernetes resource state, not individual Spark task results.</p>

<h3 style="box-sizing:border-box;font-size:18px;color:#174f83;margin:20px 0 8px;;max-width:100%;min-width:0">What changes across the three options?</h3>
<ul>
<li><strong>Standalone:</strong> Spark's Master and Worker services coordinate the Spark cluster's resources.</li>
<li><strong>YARN:</strong> the ResourceManager, NodeManagers, ApplicationMaster, and YARN containers provide resource management and application launch support.</li>
<li><strong>Kubernetes:</strong> the Control Plane schedules Pods onto nodes; Spark's Driver and Executors run in their respective Pods.</li>
</ul>
<div style="box-sizing:border-box;background:#ecfdf5;border:1px solid #a7f3d0;border-left:5px solid #10a36d;border-radius:8px;padding:14px 16px;margin-top:18px;;max-width:100%;min-width:0;overflow-wrap:anywhere">
<strong style="color:#065f46;">Core principle:</strong> The cluster manager determines how the platform manages and allocates resources. The deployment mechanism determines how application processes are launched in that environment. The Driver, through SparkContext and related runtime components, coordinates Spark execution, while Executors perform the distributed work.
